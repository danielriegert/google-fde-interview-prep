# Subsequence Problems

Whenever a problem involves subsequences and an optimal value, it's almost certainly testing dynamic programming, and the time complexity is generally O(n^2).

- [(medium)300 Longest Increasing Subsequence](https://leetcode.cn/problems/longest-increasing-subsequence/)
    Longest increasing subsequence. The constraint is relative order.
    - Application: [(hard)354 Russian Doll Envelopes](https://leetcode.cn/problems/russian-doll-envelopes/). The extra step is to sort by w ascending and by h descending.

```python
# 300
import bisect
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        len2min = [nums[0]]
        for num in nums[1:]:
            if num > len2min[-1]:
                len2min.append(num)
            elif num == len2min[-1]:
                continue
            else:
                num_id = bisect.bisect_left(len2min, num)
                len2min[num_id] = num
        return len(len2min)
```

```python
# 354
import bisect
class Solution:
    def maxEnvelopes(self, envelopes: List[List[int]]) -> int:
        envelopes.sort(key=lambda x: (x[0], -x[1]))
        len2min = [envelopes[0][1]]
        for env in envelopes:
            v = env[1]
            if v > len2min[-1]:
                len2min.append(v)
            elif v < len2min[-1]:
                insert_id = bisect.bisect_left(len2min, v)
                len2min[insert_id] = v
        return len(len2min)

```

```python
# 239: sliding window template + new sliding window design
# Runtime: 556 ms, beats 50.81% of Python3 submissions
# Memory: 28.3 MB, beats 35.06% of Python3 submissions
from collections import deque
class MaxWindow:
    def __init__(self, k):
        self.window = deque()
        self.k = k

    def push(self, n):
        while self.window and self.window[-1] < n:  # use strictly less-than here, not less-than-or-equal
            self.window.pop()
        self.window.append(n)

    def top(self):
        return self.window[0]

    def pop(self, left):
        if self.window[0] == left:
            self.window.popleft()

class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        ans = []
        max_window = MaxWindow(k)
        for i, num in enumerate(nums):
            if i < k - 1:
                max_window.push(num)
            else:
                max_window.push(num)
                ans.append(max_window.top())
                max_window.pop(nums[i-k+1])
        return ans

```

- [(medium)2128 Remove All Ones With Row and Column Flips](https://leetcode.cn/problems/remove-all-ones-with-row-and-column-flips)
    - Approach: iterate through all m+n possible operations, try each one if needed, and check at the end whether it works. A small trick: for the first number in each row, if it's 1, then it either needs to be flipped by that row, or flipped by the first column.
        - For the top-left number,
            - If it's 1, you can only choose to flip either the first row or the first column, not both. By default choose the first row, since choosing the first column is actually equivalent.
            - If it's 0, either neither the first row nor the first column is flipped, or both are flipped. If both are flipped, suppose the first row has other 0s in other columns — after flipping they become 1, so the corresponding column needs to be flipped; the same logic applies to the first column. If neither is flipped, the same logic applies. In summary, the leading element of each row and column determines whether that row/column needs to be flipped.

Selected examples + follow up


    Use the union-find idea flexibly, and get comfortable with the union-find [templates](https://books.halfrost.com/leetcode/ChapterThree/UnionFind/). There are two implementations in the template: one uses path compression + union by rank, the other tracks the size of each set + the size of the largest set. Each version has its own use cases. Problems that fit the first template: #128, #130, #547, #684, #721, #765, #778, #839, #924, #928, #947, #952, #959, #990. Problems that fit the second template: #803, #952. For #803, the rank optimization and set-size counting are time-critical — without optimizing them you will TLE.
    Union-find is really a way of thinking, and some problems require applying that idea flexibly rather than mechanically pattern-matching to the template. For example #399 is a "string union-find": it applies the union-find idea, but each node here is based on a string and a map, rather than being implemented purely with integer node IDs.
    For some problems, blindly applying the template actually fails, e.g. #685. This problem cannot use path compression or union by rank, because it involves a directed graph and we need to know a node's predecessor — if the path were compressed, this problem couldn't be solved. So this problem must NOT use path compression or union by rank.
    Flexibly abstract the information given in a problem, assign IDs to the given information in a sensible way, then solve with union-find, using a map to reduce time complexity — see #721 and #959.
    For problems about maps, bricks, or grids, you can create one special "virtual" node and union() all the border bricks/cells into that special node. See #130 and #803.
    Problems solvable with union-find can generally also be solved with DFS and BFS, just with somewhat higher time complexity.

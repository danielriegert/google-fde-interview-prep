# Implementation of various sorting algorithms

```python
# Bubble Sort
def BubbleSort(nums):
    # by default, sort by ascending order
    for i in range(1, len(nums)):
        swap = False
        for j in range(len(nums) - i):
            if nums[j+1] < nums[j]:
                nums[j+1], nums[j] = nums[j], nums[j+1]
                swap = True
        if not swap: # for early stop, to reduce the time complexity for the best case.
            break
```

```python
nums = [1,4,6,2,4,2,1]
BubbleSort(nums)
print(nums)
```

```python
# Selection Sort
def SelectionSort(nums):
    # by default, sort by ascending order
    for i in range(len(nums)):
        min_id = -1
        for j in range(i, len(nums)):
            if min_id < 0 or nums[j] < nums[min_id]:
                min_id = j
        nums[i], nums[min_id] = nums[min_id], nums[i]
```

```python
nums = [1,4,6,2,4,2,1]
SelectionSort(nums)
print(nums)
```

```python
# Insertion Sort
def InsertionSort(nums):
    # by default, sort by ascending order
    for i in range(1, len(nums)):
        tmp = nums[i]
        j = i - 1
        while j >= 0 and tmp < nums[j]:
            nums[j+1] = nums[j]
            j -= 1
        nums[j + 1] = tmp
```

```python
nums = [3,6,2,4,2,1]
InsertionSort(nums)
print(nums)
```

```python
# Merge Sort recursive implementation, requires extra O(n) space
def MergeSort(nums):
    if len(nums) > 1:
        middle = len(nums) // 2
        return merge(MergeSort(nums[:middle]), MergeSort(nums[middle:]))
    return nums

def merge(nums1, nums2):
    new_nums = []
    while nums1 and nums2:
        if nums1[0] <= nums2[0]:
            new_nums.append(nums1.pop(0))
            
        else:
            new_nums.append(nums2.pop(0))
    if nums1:
        new_nums.extend(nums1)
    else:
        new_nums.extend(nums2)
    return new_nums
    
```

```python
import heapq
def HeapSort(nums):
    heap = []
    for num in nums:
        heapq.heappush(heap, num)
    nums = []
    while heap:
        nums.append(heapq.heappop(heap))
    return nums
        
```

```python
nums = [3,6,2,4,2,5,7,1]
nums = HeapSort(nums)
print(nums)
```

```python
def QuickSort(nums):
    pivot = 0
    for i in range(1, len(nums)):
        if nums[i] < nums[0]:
            pivot += 1
            nums[pivot], nums[i] = nums[i], nums[pivot]
    print(nums)
    if pivot > 0:
        nums[0], nums[pivot] = nums[pivot], nums[0]
        nums[:pivot] = QuickSort(nums[:pivot])
    if pivot < len(nums)-1:
        nums[pivot + 1: ] = QuickSort(nums[pivot + 1: ])
    return nums
```

```python
nums = [3,6,2,4,2,5,7,1]
nums = QuickSort(nums)
print(nums)
```

```python
# Improved version: added an early stop, reducing the time complexity for already-sorted input to O(n)
def QuickSort2(nums):
    pivot = 0
    changed = False
    for i in range(1, len(nums)):
        if nums[i] < nums[0]:
            changed = True
            pivot += 1
            nums[pivot], nums[i] = nums[i], nums[pivot]
    print(nums)
    if not changed:
        return nums
    if pivot > 0:
        nums[0], nums[pivot] = nums[pivot], nums[0]
        nums[:pivot] = QuickSort(nums[:pivot])
    if pivot < len(nums)-1:
        nums[pivot + 1: ] = QuickSort(nums[pivot + 1: ])
    return nums
```

```python
# Bucket Sort
from collections import Counter
def BucketSort(nums):
    if not nums:
        return nums
    min_val, max_val = min(nums), max(nums)    
    buckets = Counter(nums)
    res = []
    for i in range(min_val, max_val + 1):
        if i in buckets:
            res.extend([i] * buckets[i])
    return res
    
```

```python
nums = [3,6,12,4,25,5,47,1]
nums = BucketSort(nums)
print(nums)
```

```python
def RadixSort(nums):
    k = len(str(max(nums)))
    for j in range(k):
        buckets = [[] for i in range(10)]
        for num in nums:
            buckets[num//(10**j)%10].append(num)
        nums = []
        for b in buckets:
            nums.extend(b)
        print(j, nums)
    return nums
```

```python
nums = [3,6,12,4,25,5,47,1]
nums = RadixSort(nums)
print(nums)
```

```python
# 75 Quicksort
class Solution:
    def quicksort(self, nums, start_id, end_id):
        if start_id < end_id - 1:
            index = start_id
            for i in range(start_id + 1, end_id):
                num = nums[i]
                if num < nums[start_id]:
                    index += 1 # Detail: increment index first, then swap. Otherwise index ends up one more than expected when the loop exits
                    nums[index], nums[i] = num, nums[index]
            if index > start_id:
                tmp = nums[start_id]
                nums[start_id] = nums[index]
                nums[index] = tmp
            self.quicksort(nums, start_id, index)
            self.quicksort(nums, index + 1, end_id)


    def sortColors(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        self.quicksort(nums, 0, len(nums))
```

```python
# 148 Merge sort, accepted
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head or not head.next: # Detail 1: if it's None or has only one node, return directly
            return head
        faster = head
        slower_head = ListNode(next=head) # Detail 2: this actually points to the node before the slow pointer. E.g. with 2 nodes, after the first loop iteration the fast pointer reaches the end, the slow pointer reaches the second node, and the node pointing to the slow pointer reaches the first node.
        ordered = True # Detail 3: for early stop

        while faster and faster.next:
            if slower_head.next.next.val < slower_head.next.val:
                ordered = False
            faster = faster.next.next
            slower_head = slower_head.next
        
        mid = slower_head.next # start node of the second half
        slower_head.next = None
        if not ordered:
            head = self.sortList(head) # corresponds to Detail 3: if one half is already sorted, skip one round of sorting
        next_head = self.sortList(mid)
        dummy_head = ListNode() # Detail 4: dummy head pointer, to keep the loop format uniform
        cur_head = dummy_head
        while head and next_head:
            if head.val <= next_head.val:
                cur_head.next = head
                head = head.next
            else:
                cur_head.next = next_head
                next_head = next_head.next
            cur_head = cur_head.next
        if head:
            cur_head.next = head
        else:
            cur_head.next = next_head
        return dummy_head.next
```

```python
# 148 Bucket sort, accepted
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return head
        bucket = {}
        min_val = head.val
        max_val = head.val
        cur = head
        while cur:
            bucket[cur.val] = bucket.get(cur.val, 0) + 1
            min_val = min(min_val, cur.val)
            max_val = max(max_val, cur.val)
            cur = cur.next
        cur = head
        for i in range(min_val, max_val + 1):
            if i in bucket:
                for _ in range(bucket[i]):
                    cur.val = i
                    cur = cur.next
        return head
```

```python
# 937
class Solution:
    def reorderLogFiles(self, logs: List[str]) -> List[str]:
        def trans(log):
            a, b = log.split(" ", 1)
            return (0, b, a) if b[0].isalpha() else (1,)
        logs.sort(key=trans)
        return logs
```

```python
# 179 Approach 1: Bucket sort

class Solution:
    def largestNumber(self, nums: List[int]) -> str:
        def bucketsort(nums, default_val=""):
            if len(nums) == 1:
                return default_val + nums[0]
            buckets = {}
            empty_cnt = 0
            for num in nums:
                if num == "":
                    empty_cnt += 1
                    continue
                start_digit = num[0]
                if start_digit not in buckets:
                    buckets[start_digit] = [num[1:]]
                else:
                    buckets[start_digit].append(num[1:])
            merge_strs = {}
            for key, val in buckets.items():
                merge_strs[key] = bucketsort(val, default_val + key)
            merged_str = ""
            for i in range(9, -1, -1):
                if str(i) not in merge_strs:
                    continue
                if default_val != "" and str(i) == default_val[0]: # special handling for the sub-bucket whose key equals the prefix's first character
                    if default_val + merge_strs[str(i)] > merge_strs[str(i)] + default_val: 
                        merged_str += ("".join([default_val for _ in range(empty_cnt)]))
                        merged_str += merge_strs[str(i)]
                        empty_cnt = 0
                    else:
                        merged_str += merge_strs[str(i)]
                        merged_str += ("".join([default_val for _ in range(empty_cnt)]))
                        empty_cnt = 0
                else:
                    if (default_val == "" or i < int(default_val[0])) and empty_cnt > 0:
                        merged_str += ("".join([default_val for _ in range(empty_cnt)]))
                        empty_cnt = 0
                    merged_str += merge_strs[str(i)]
            if empty_cnt > 0: # special handling for the '' (empty string) case
                merged_str += ("".join([default_val for _ in range(empty_cnt)]))
            return merged_str
        uncleand = bucketsort([str(num) for num in nums], default_val="")
        for i in range(len(uncleand)): # clean up the final output
            if uncleand[i] != '0':
                break
        return uncleand[i:]
```

```python
# 179 Approach 2: Rewrite the sort comparator
import functools
class Solution:
    def largestNumber(self, nums: List[int]) -> str:
        def cmp(a, b):
            if a + b == b + a:
                return 0
            if a + b > b + a:
                return -1
            return 1
        str_nums = [str(item) for item in nums]
        str_nums.sort(key=functools.cmp_to_key(cmp))
        res = ''.join(str_nums)
        for i in range(len(res)):
            if res[i] != '0':
                break
        return res[i:]
```

```python
# 215 Approach: Quickselect to find the pivot, no need to fully sort
import random
class Solution:
    def fastsort(self, nums, start_id, end_id, aim_id):
        if end_id == start_id:
            return
        pivot_id = random.randint(start_id, end_id) # randomly choose the pivot
        if pivot_id != start_id:
            nums[start_id], nums[pivot_id] = nums[pivot_id], nums[start_id] # put the pivot at the first position, to make the loop easier
        swap_id = start_id
        for i in range(start_id +1, end_id + 1):
            if nums[i] <= nums[start_id]:
                swap_id += 1
                nums[swap_id], nums[i] = nums[i], nums[swap_id]
        nums[start_id], nums[swap_id] = nums[swap_id], nums[start_id]
        # determine the relative position
        if swap_id == aim_id:
            return
        if swap_id < aim_id:
            self.fastsort(nums, swap_id + 1, end_id, aim_id)
        else:
            self.fastsort(nums, start_id, swap_id - 1, aim_id)

    def findKthLargest(self, nums: List[int], k: int) -> int:
        k_smaller_id = len(nums) - k
        self.fastsort(nums, 0, len(nums) - 1, k_smaller_id)
        return nums[k_smaller_id]
```

```python
# 220 Approach 2: Bucket sort
class Solution:
    def containsNearbyAlmostDuplicate(self, nums: List[int], k: int, t: int) -> bool:
        buckets = {}
        for i, numi in enumerate(nums):
            bucket_id = numi // (t + 1) # bucket partition scheme: each bucket spans t+1 consecutive values, which guarantees that values within a bucket satisfy the condition.
            if bucket_id in buckets:
                return True
            buckets[bucket_id] = numi
            if bucket_id - 1 in buckets: # need to, and only need to, check the two adjacent buckets
                if numi - buckets[bucket_id - 1] <= t:
                    return True
            if bucket_id + 1 in buckets:
                if buckets[bucket_id + 1] - numi <= t:
                    return True
            if i > k: # remove elements that have fallen outside the window from their bucket
                buckets.pop(nums[i-k] // (t+1))
        return False
```

```python
# 324 Bucket sort

class Solution:
    def wiggleSort(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        buckets = {}
        for num in nums:
            buckets[num] = buckets.get(num, 0) + 1
        ttl_len = len(nums)
        if ttl_len % 2 == 1:
            small_end = ttl_len
            big_end = ttl_len - 1
        else:
            big_end = ttl_len
            small_end = ttl_len - 1
        start_num = 5000
        for i in range(1, big_end, 2):
            while start_num not in buckets or buckets[start_num] == 0:
                start_num -= 1
            nums[i] = start_num
            buckets[start_num] -= 1
        for i in range(0, small_end, 2):
            while start_num not in buckets or buckets[start_num] == 0:
                start_num -= 1
            nums[i] = start_num
            buckets[start_num] -= 1
        return nums
```

```python
# 451  Approach 2: Find the max frequency, use bucket sort
from collections import Counter
class Solution:
    def frequencySort(self, s: str) -> str:
        alpha_cnt = Counter([c for c in s])
        buckets = {}
        max_buckets = 0
        for key, val in alpha_cnt.items():
            if val in buckets:
                buckets[val].append(key)
            else:
                buckets[val] = [key]
            max_buckets = max(max_buckets, val)
        res = ""
        cur_cnt = max_buckets
        while cur_cnt > 0:
            if cur_cnt in buckets:
                for alpha in buckets[cur_cnt]:
                    res += alpha * cur_cnt
            cur_cnt -= 1
        return res
```

```python
# 767 Bucket sort + max-heap, fill two letters at a time.
from collections import Counter
import heapq
class Solution:
    def reorganizeString(self, s: str) -> str:
        alpha_cnt = Counter([c for c in s])
        max_heap = []
        for alpha, cnt in alpha_cnt.items():
            heapq.heappush(max_heap, (-cnt, alpha))
        res = ""
        while len(max_heap) > 1:
            cur_cnt, cur_alpha = heapq.heappop(max_heap) # each time, take the two letters with the most remaining occurrences and place them side by side.
            next_cnt, next_alpha = heapq.heappop(max_heap)
            res += (cur_alpha + next_alpha)
            cur_cnt += 1
            next_cnt += 1
            if cur_cnt < 0:
                heapq.heappush(max_heap, (cur_cnt, cur_alpha)) # push the remaining count back onto the heap
            if next_cnt < 0:
                heapq.heappush(max_heap, (next_cnt, next_alpha))
        if max_heap:
            final_cnt, final_alpha = heapq.heappop(max_heap)
            if final_cnt < -1:
                return ''
            res += final_alpha
        return res
```

```python
# 767 Approach 2: Fill even positions first, then odd positions

from collections import Counter
import heapq
class Solution:
    def reorganizeString(self, s: str) -> str:
        res = [c for c in s]
        alpha_cnt = Counter(res)
        max_heap = []
        for alpha, cnt in alpha_cnt.items():
            heapq.heappush(max_heap, (-cnt, alpha))
        
        start_id = 0 # fill even positions first
        max_cnt, _ = max_heap[0]
        if -max_cnt > (len(res) + 1) // 2: # if the most frequent letter's count exceeds (len(res) + 1) // 2, it's impossible; otherwise it's feasible.
            return ""
        while max_heap:
            cur_cnt, cur_alpha = heapq.heappop(max_heap)
            while cur_cnt < 0 and start_id < len(s):
                res[start_id] = cur_alpha
                start_id += 2
                cur_cnt += 1
                if start_id >= len(s):
                    start_id = 1 # then fill odd positions
        return "".join(res)
```

```python
# 1054  Same as 767
import heapq
from collections import Counter
class Solution:
    def rearrangeBarcodes(self, barcodes: List[int]) -> List[int]:
        barcodes_cnt = Counter(barcodes)
        res = [0 for bar in barcodes]
        start_id = 0
        max_heap = []
        for barcode, cnt in barcodes_cnt.items():
            heapq.heappush(max_heap, (-cnt, barcode))
        while max_heap:
            cur_cnt, cur_barcode = heapq.heappop(max_heap)
            while cur_cnt < 0:
                res[start_id] = cur_barcode
                cur_cnt += 1
                start_id += 2
                if start_id >= len(barcodes):
                    start_id = 1
        return res
```

```python
# 969 Sort from tail to head, one at a time. Put the largest element at the very end.
class Solution:
    def flip(self, arr, k):
        for i in range((k+1)//2):
            arr[i], arr[k-i-1] = arr[k-i-1], arr[i]
        
    def pancakeSort(self, arr: List[int]) -> List[int]:
        end = len(arr) # end_id + 1, from tail to head, put the largest element at the very end.
        res = []
        while end > 1:
            end -= 1
            max_id = 0
            max_val = arr[0]
            change_times = 0 # track how many times the max value changed, used for early stop on an already-sorted array
            for i in range(end + 1):
                if arr[i] > max_val:
                    max_val = arr[i]
                    max_id = i
                    change_times += 1
            if change_times == end + 1: # early stop for an already-sorted array
                break
            if max_id == end:
                continue
            self.flip(arr, max_id + 1)
            self.flip(arr, end + 1)
            res.extend([max_id + 1, end + 1])
        return res
```

```python
# 164 Bucket sort
class Solution:
    def maximumGap(self, nums: List[int]) -> int:
        min_val = min(nums)
        max_val = max(nums)
        if len(nums) == 1 or max_val == min_val:
            return 0
        bucket_dist = int((max_val - min_val) / (len(nums) - 1)) # the smallest possible value of the max gap; technically should be rounded up
        if bucket_dist == 0:
            bucket_dist = 1
        buckets = {}
        max_gap = 0
        for num in nums:
            bucket_id = num // bucket_dist
            if bucket_id in buckets:
                # each bucket only needs to store two values: the max and the min within that bucket
                buckets[bucket_id] = [min(buckets[bucket_id][0], num), max(buckets[bucket_id][1], num)]
            else:
                buckets[bucket_id] = [num, num]
        last_max = -1
        for bucket_id in range(min_val // bucket_dist, max_val // bucket_dist + 1):
            if bucket_id in buckets:
                if last_max > -1:
                    # subtracting the current bucket's max from the next bucket's min gives the gap between the two adjacent numbers that fall in these two buckets
                    max_gap = max(max_gap, buckets[bucket_id][0] - last_max)
                last_max = buckets[bucket_id][1]
        return max_gap
```

# My Summary

There are two ways to sort in Python: `sorted()` and `list.sort()`
- `sorted(list, func)`: returns the sorted list; `func` lets you customize the comparison. The way it works: pass in two elements to compare, x and y,
    - if x should come before y, return -1,
    - if x should come after y, return 1.
    - if x and y are equal, return 0. (presumably to preserve stability)
    
-  `list.sort(cmp=none, key=none, reverse=none)` (python2), `list.sort(key=none, reverse=none)` (python3). Here, `cmp` is a customizable comparison function. In python3, a custom comparison function can be defined using the following method

- There are many implementation details to be careful about when implementing sorting algorithms.
    - Quicksort: increment index first, then swap. Otherwise, index ends up one more than expected when the loop exits
    
- Rewriting the sort comparator is a very useful problem-solving technique. See problems 937, 179

- Bucket sort can, to some extent, reduce sorting from O(nlogn) complexity down to linear.
    - Sometimes the tricky part is how to design the buckets. See problems 164, 220
    
- For problems like wiggle sort, the key insight is often what index pattern to follow when inserting data. See problems 324, 767

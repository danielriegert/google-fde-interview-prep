    - Approach 2: the approach above can time out — since the number can be as large as 10^10, anything with N or log N complexity will time out. It turns out there's an O(1) closed-form solution. The idea (after first dividing everything by 2, i.e. working with odd numbers) is to use the Gauss sum formula to partition 0..N: if N exactly equals the sum 1+...+j, the max count is j; otherwise you can always express it using j-1 numbers, where j is the position of the first Gauss sum greater than N. - [(easy)359 Logger Rate Limiter](https://leetcode.cn/problems/logger-rate-limiter)
    [come from](https://www.cnblogs.com/grandyang/p/5592635.html)
Design a logger system that receive stream of messages along with its timestamps, each message should be printed if and only if it is not printed in the last 10 seconds.

Given a message and a timestamp (in seconds granularity), return true if the message should be printed in the given timestamp, otherwise returns false.

It is possible that several messages arrive roughly at the same time.
    - Approach: at first glance this looks like a sliding-window problem, but it's actually a hash map (a lazy-update version of a sliding window). Key is the message, value is the timestamp. To decide whether to print at a given timestamp, look up the hash map. If the message isn't present, return true. If it is present and the timestamp gap is greater than 10, return true; otherwise return false. Then set the value to the new timestamp. Space complexity: grows with the number of distinct messages, which keeps increasing — messages can be cleaned up periodically.
    - Approach 2: sliding window — keep moving the window forward based on the timestamp, using a counter to track which messages are in the window and how many times each occurs. Each time a message slides out of the window, update its counter. On a query, check whether the message is in the counter: if not, return true and add the message to the counter; otherwise return false. Space complexity: an array sized to the window length, plus a counter whose size can also be on the order of the window length.

- Given a set of logs with task id, timestamp, and start/end event, plus a timeout value, find the first task that times out. Not every task id that has a "start" log also has a matching "end" log. Feels like I've seen a similar problem before but can't remember the number. Follow-up: what if the timeout isn't constant, and what if the input is a stream?


In the initial version, the timeout is fixed (e.g. 4) and the same for every log; the logs are already sorted by time, so no extra sorting is needed — just a queue to check for timeouts and a hashset to track tasks that haven't ended yet. If a task that has timed out (per the queue) is still in the hashset, that's the answer.
For example, given input (id, timestamp, event), timeout = 3
(1, 0, start) (2,1,start) (1,2,end) (2,8,end) (3,10,start)
task2 is the first one to time out, and the goal is to detect it as early as possible; a task like task3 that never gets an "end" log also counts as timed out. In the follow-up, the timeout differs per task — just use a heap instead of a queue, and this approach still works when the input is a stream.

- Given a set of folders, each with an id, name, and parent, print the path from each folder up to the root. Need to detect cycles, use memoization to optimize the complexity, and handle various edge cases

The input looks roughly like this: (id, displayName, parent)
folder(1,"folder1",-1)
folder(2,"folder2",1)
folder(4,"folder4",2)
folder(3,"folder3",-1)
The output would be "folder1","folder1/folder2","folder1/folder2/folder4","folder3"
The id is guaranteed unique; whether the name is unique doesn't matter. The root's parent is -1. Brute force is to build a graph and then, for every folder, walk up to the root — O(n^2). Using **recursive DFS + memoization** gets it down to O(n): cache each folder's resolved path, e.g. f1->f2->f3->...->f100 — if f2 has already been computed, then when resolving f1 you can reuse it directly instead of re-traversing; if not, recurse into f2. Same idea as top-down DP, so every folder is only visited once.

First question: a beefed-up version of #278, find all the bad commits. I've seen this problem mentioned on 1point3acres before, but never saw anyone explain the problem statement clearly, so I'll go into detail here. In short, the input is two integers, low and high, representing the first and last commit. You also have an API isWorse(a, b) that tells you whether commit b is worse than commit a. Essentially, you're given a sequence that only gets worse over time (e.g. conceptually like the monotonically non-increasing digits of 999887776555322211), and you need to find every index that is worse than the previous one — except the indices don't start at 0, they range from low to high, and the indices in between are contiguous. Just use binary search: within a search range, if the right endpoint isn't worse than the left endpoint, that range definitely has no worse commits, so return immediately. Otherwise, take the midpoint, check whether it's a bad commit — if so, add it to the answer — and recurse into binary search on both halves

Follow-up: unlimited bombs — what is the minimum number needed to reach the endpoint? Solve with Dijkstra: use a 2D array to store each cell's optimal k (bombs used), initialized to infinity, with the start cell's k = 0. Then queue = [(0, start)]; each time you heap-pop a point from the queue, check whether it can reduce the optimal k of its 4 neighbors, and if so, heap-push those into the queue. Once the queue is empty, the endpoint's optimal k is the answer

If not all treasures can be collected, return -1
Follow-up: the treasures are now locked. Lowercase letters are keys, and lowercase/uppercase pairs correspond to each other — e.g. key 'a' unlocks treasure 'A'. Still find the shortest path to the endpoint, still need to collect all the treasures, but collecting a treasure now requires finding its key first
Both questions follow a similar idea — the basic BFS pattern, tracking not just visited cells but also which treasures/keys have already been collected. I used a bitmask for this, since there are at most 26 treasures/keys, which fits neatly into a single int representing all states (probably the intent of the problem design). Also, the treasure letters aren't necessarily contiguous — you could have A and C but no B — and a bitmask handles that fine. When you encounter a treasure, check whether you have the matching key; and when you reach the endpoint, check whether all treasures have been collected.

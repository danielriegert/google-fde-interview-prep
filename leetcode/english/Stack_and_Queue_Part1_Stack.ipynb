```python
# Stack implemented with an array

class Stack(object):
    def __init__(self):
        self.vec = []
        
    def push(self, num):
        self.vec.append(num)
        
    def isEmpty(self):
        return len(self.vec) == 0
    
    def top(self):
        if self.isEmpty():
            return None
        return self.vec[-1]
        
    def pop(self):
        if self.isEmpty():
            return None
        else:
            last_item = self.top()
            self.vec = self.vec[:-1] # this step can be optimized to self.vec.pop(), see pop_v2
            return last_item
        
    def pop_v2(self):
        if self.isEmpty():
            return None
        else:
            last_item = self.top()
            self.vec.pop()
            return last_item
```

```python
s = Stack()

queue = [1, 2, 93, 4, 3, 5]
for i in queue:
    s.push(i)
    print("top: ", s.top())
while not s.isEmpty():
    print("pop: ", s.pop())
```

```python
s = Stack()

queue = [1, 2, 93, 4, 3, 5]
for i in queue:
    s.push(i)
    print("top: ", s.top())
while not s.isEmpty():
    print("pop: ", s.pop_v2())
```

```python
# Stack implemented with a linked list

class Node(object):
    def __init__(self, val):
        self.val = val
        self.next = None
        
    def setNext(self, node):
        self.next = node  
    
    def getNext(self):
        return self.next
        
    def getVal(self):
        return self.val

class Stack(object):
    def __init__(self):
        self.head = None
        
    def push(self, num):
        next_node = Node(num)
        next_node.setNext(self.head)
        self.head = next_node
        
    def isEmpty(self):
        return self.head is None
    
    def top(self):
        if self.isEmpty():
            return None
        return self.head.getVal()
        
    def pop(self):
        if self.isEmpty():
            return None
        else:
            last_item = self.top()
            self.head = self.head.getNext()
            return last_item
        
```

```python
s = Stack()

queue = [1, 2, 93, 4, 3, 5]
for i in queue:
    s.push(i)
    print("top: ", s.top())
while not s.isEmpty():
    print("pop: ", s.pop())
```

# Time and Space Complexity

Stack:

Array implementation: time complexity
- Add: O(1)
- Delete: O(1)

Space complexity O(N)

Implementing a stack with a linked list has no significant advantage. The advantage of a linked list is that adding and removing elements is easy, but for a stack, both adding and removing happen at the top, so using an array directly already gives O(1) — there's no need to use a linked list.

When the popped value equals the current minimum, pop the min stack too.         - Most direct idea: for each bar, compute the max area of a rectangle with that bar as its rightmost edge. Iterate over all bars and find the overall maximum. Problem: O(n^2) complexity, likely times out.
        - Monotonic stack idea: push the height of a "blocking" bar along with its corresponding max area. Definition of a blocking bar: the corresponding max area is limited by that bar's height.     - [(hard) 726 Number of Atoms](https://leetcode-cn.com/problems/number-of-atoms/). Approach: because of the brackets, the most direct idea is recursion. When the formula is an atomic expression, stop recursing and compute the count of each element. Otherwise, split the formula and evaluate each part. The difficulty of this problem is the large number of details that need to be handled.

```python
# 42
class Solution:
    def calcVolum(self, stack): # perform the operation to compute the trapped water volume.
        vol = 0
        if len(stack) == 0:
            return 0
        right_edge = stack[-1]
        for hi in range(len(stack) - 2, 0, -1):
            right_edge = max(right_edge, stack[hi + 1])
            edge = min(stack[0], right_edge) # the water each bar can hold equals the height difference from the shorter of the two sides. Each side's height is determined by the tallest bar on that side. The tallest bar on the left is the left endpoint, determined by push order. The tallest bar on the right needs to be determined iteratively.
            if stack[hi] < edge:
                vol += edge - stack[hi]
        return vol

    def trap(self, height: List[int]) -> int:
        stack = []
        res = 0
        for h in height:
            if len(stack) == 0: # if the stack is empty and the height is 0, don't push.
                if h > 0: stack.append(h)
                continue
            stack.append(h) # push
            if h >= stack[0]: # compute the trapped water volume
                res += self.calcVolum(stack)
                stack = [h] # after computing, clear the stack. Push this bar again as the new left edge.
        res += self.calcVolum(stack) # compute the trapped water volume
        return res
```

```python
# 071

class Solution:
    def stack2str(self, stack): # output this stack
        if len(stack) == 0:
            return "/"
        res = ""
        for item in stack:
            res += "/%s" % item
        return res

    def simplifyPath(self, path: str) -> str:
        items = path.split("/")
        stack = []
        for item in items:
            # how to operate stack according to the item
            if item in ["", "."]:
                continue
            if "/" in item:
                continue
            if item == "..":
                if len(stack) > 0:
                    stack.pop()
            else:
                stack.append(item)
    
        return self.stack2str(stack)
```

```python
# 321 
class Solution:
    def maxNumberSingle(self, num, k):
        removed_k = len(num) - k
        res = []
        for n in num:
            while removed_k > 0 and res and res[-1] < n:
                res.pop()
                removed_k -= 1
            res.append(n)
        return res[:k]

    def combine(self, nums1, nums2):
        res = []
        while nums1 or nums2:
            bigger = nums1 if nums1 > nums2 else nums2 # trick: Python list comparison
            res.append(bigger[0])
            bigger.pop(0)
        return res

    def maxNumber(self, nums1: List[int], nums2: List[int], k: int) -> List[int]:
        global_max = []
        for k1 in range(max(0, k - len(nums2)), min(k, len(nums1)) + 1): # iterate over all possible splits, take the maximum result
            k2 = k - k1
            max1 = self.maxNumberSingle(nums1, k1) # apply 402 to a single array
            max2 = self.maxNumberSingle(nums2, k2)
            local_max = self.combine(max1, max2) # merge the results of the two arrays into the lexicographically largest combination
            global_max = global_max if global_max > local_max else local_max # trick: Python list comparison
        return global_max
```

```python
# 456 approach 2: improved version
class Solution:
    def find132pattern(self, nums: List[int]) -> bool:
        minimum = nums[0]
        cnt = len(nums)
        max_oracle = [[False, i] for i in range(cnt)]
        min_val_oracle = []
        smaller_stack = []
        # reverse iteration to find the larger number and the index of the nearest larger number.
        for i in range(cnt):
            last_id = cnt - i - 1
            while smaller_stack and smaller_stack[-1][0] < nums[last_id]:
                max_oracle[smaller_stack[-1][1]] = [True, last_id] # when the current number is greater than the number in the stack, pop, and save the record for the stacked number (which comes after the current number)
                smaller_stack.pop()
            smaller_stack.append([nums[last_id], last_id])

        # forward iteration to find the minimum.
        for i, n in enumerate(nums):
            if minimum > n:
                minimum = n
            min_val_oracle.append(minimum)
        
        # reverse iteration: find a k that satisfies the condition
        for i in range(cnt):
            k = cnt - i - 1
            if max_oracle[k][0]:
                j = max_oracle[k][1]
                if min_val_oracle[j] < nums[k]:
                    return True
        return False
```

```python
# 739
class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        stack = [] 
        answer = [0 for _ in temperatures] # default answer is 0
        for i in range(len(temperatures)- 1, -1, -1):
            if not stack:
                stack.append([temperatures[i], i])
            else:
                while stack and stack[-1][0] <= temperatures[i]: # if the temperature in the stack is lower than or equal to the current temperature, it's useless for earlier dates and can be popped.
                    stack.pop()
                if stack:
                    answer[i] = stack[-1][1] - i
                stack.append([temperatures[i], i])
        return answer
```

```python
# 901
class StockSpanner:

    def __init__(self):
        self.stack = []
        self.day = 0


    def next(self, price: int) -> int:
        self.day += 1
        while self.stack and self.stack[-1][0] <= price:
            self.stack.pop()
        last_day = 0
        if self.stack:
            last_day = self.stack[-1][1]
        self.stack.append([price, self.day])
        return self.day - last_day
```

```python
# 907
class Solution:
    def sumSubarrayMins(self, arr: List[int]) -> int:
        stack = [] # item [val, index, end_sum] — maintains elements less than or equal to the current position, their index, and the corresponding subarray sum.
        ttl_sum = 0
        for i, val in enumerate(arr):
            while stack and stack[-1][0] > val: # if the top of the stack is greater than the current element, pop it, until the top is less than or equal to the current element.
                stack.pop()
            if not stack:
                cur_sum = val * (i + 1) # compute the sum of minimums over subarrays ending at the current element. If the stack has no other elements, that proves all previous elements are greater, so the minimum of all i+1 subarrays is the current element.
            else:
                cur_sum = val * (i - stack[-1][1]) + stack[-1][2] # otherwise, the sum of minimums equals the sum of subarrays where the current element is the minimum, plus the sum for all subarrays ending at the previous smaller element.
            stack.append([val, i, cur_sum])
            ttl_sum += cur_sum # total: add the sum of minimums over all subarrays ending at the current element.
        return ttl_sum % (10 ** 9 + 7)
```

```python
#2104  approach 1: brute force
class Solution:
    def subArrayRanges(self, nums: List[int]) -> int:
        res = 0
        for i in range(len(nums)):
            min_val = nums[i]
            max_val = nums[i]
            for j in range(i + 1, len(nums)):
                min_val = min(min_val, nums[j])
                max_val = max(max_val, nums[j])
                res += (max_val - min_val)
        return res
```

```python
#2104  approach 2: monotonic stack
class Solution:
    def subArrayRanges(self, nums: List[int]) -> int:
        res = 0
        min_stack = [] # id, sum
        max_stack = []
        minus_min = 0 # tracks the sum of minimums over all subarrays ending at the current element
        add_max = 0 # tracks the sum of maximums over all subarrays ending at the current element
        for i, val in enumerate(nums):
            while min_stack and nums[min_stack[-1][0]] > val: # if the top of the stack is greater than the current element, pop it, based on "forward closure".
                _, out_sum = min_stack.pop()
                minus_min -= out_sum # also remove its contribution from the sum of minimums.
            if not min_stack:
                last_id = -1 # if the stack is empty, use -1, so the index maps correctly to the count.
            else:
                last_id, _ = min_stack[-1] # if the stack has elements, the scope of the current "minimum" (the starting index of subarrays where it's the minimum) begins right after the top element's index.
            min_stack.append([i, val * (i - last_id)]) # the stack maintains the sum over the scope's subarrays, to make computation easier.
            minus_min += min_stack[-1][1]

            while max_stack and nums[max_stack[-1][0]] <= val: # the same logic applies to the max stack
                _, out_sum = max_stack.pop()
                add_max -= out_sum
            if not max_stack:
                last_id = -1
            else:
                last_id, _ = max_stack[-1]
            max_stack.append([i, val * (i - last_id)])
            add_max += max_stack[-1][1] 
            res += (add_max - minus_min) # the sum over all subarrays equals the sum of maximums minus the sum of minimums
        return res
```

# Summary / Takeaways

Stacks are mainly used for bracket-related problems.

For stacks, especially monotonic stacks, using them well can significantly reduce time complexity. But how do you spot when a stack should be used?
- The most useful method is to first **brute-force** a solution, then **observe** what repeated computations are happening and what characterizes them. Can a stack maintain that pattern?
- Also look for mathematical patterns, e.g. the pattern in problem 456 (132 pattern), the minimal-order property in 402, and the property of the unsorted subarray in 581.

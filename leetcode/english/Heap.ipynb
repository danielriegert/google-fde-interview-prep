# Basic Concepts

A heap is defined as a structure satisfying the following two conditions
- Complete binary tree: a binary tree where every level is full except possibly the last
- The value of any node is always greater than or equal to (or less than or equal to) its children, corresponding to a max-heap or min-heap respectively

Heap operations
- Insert: insert a new element into the heap
- Delete: remove the top element of the heap
- Build heap: turn an array into a heap
- top: view the top element of the heap

# Heap Implementation
Since a heap is a complete binary tree, this means we don't actually need to use a real tree structure to implement it — we can implement it directly with an **array**, and get the parent/child relationships by computing indices.

- Parent/child index relationship: assuming indices start at 0, if a parent node's index is n, its two children's indices are 2n+1 and 2n+2. Its parent's index is (n-1)//2.
    - Derivation method 1 (induction): for the node at index 0, its two children's indices are 1 and 2, which satisfies the rule; assume the node at index n satisfies this rule, then for the node at index n+1, its children are (2n+2)+1=2(n+1)+1 and (2n+2)+2=2(n+1)+2. By induction, this holds.
    - Derivation method 2: suppose the node at index n is the k-th node on level l. Then its children are the (2(k-1)+1)-th and (2(k-1)+2)-th nodes on level l+1. How many nodes are there in total across levels 1 through l? 2^l-1. So for the node at index n, we have n + 1 = 2^(l-1)-1+k => 2n = 2^l + 2k-4, and its children's indices are 2^l - 1 + 2(k-1)+1 - 1 = 2^l+2k-2 - 1 = 2n+4-2 = 2n+1, and 2^l - 1 + 2(k-1)+2 - 1 = 2^l + 2k - 2 = 2n+2.
    - Extension: if indices start at 1, then if a parent's index is n, its two children's indices are 2(n-1)+1+1=2n and 2(n-1)+2+1=2n+1, and the inverse relationship is n//2
    
  
- [X] Can this be implemented with a linked list? Sure — but would the insert/delete complexity be any smaller? No, you'd still need to do repeated comparisons.

```python
# Array-based heap implementation, min-heap
class MinHeap(object):
    def __init__(self, vec=[]):
        self.vec = []
        self.len = 0
        if vec:
            self.heapify(vec)
            
    def push(self, data):
        self.vec.append(data) # first place it at the very end
        self.len += 1
        child_id = self.len - 1
        father_id = (child_id - 1) // 2
        while father_id >= 0 and self.vec[child_id] < self.vec[father_id]: # sift-up operation, swap with the larger parent
            self.vec[father_id], self.vec[child_id] = self.vec[child_id], self.vec[father_id]
            child_id = father_id
            father_id = (child_id - 1) // 2
        
    def isEmpty(self):
        return self.len == 0
    
    
    def top(self):
        if self.isEmpty():
            return False
        return self.vec[0]
    
    def pop(self):
        if self.isEmpty():
            return False
        top_val = self.vec[0] # remove the top element of the heap
        self.vec[0] = self.vec[-1]  # move the last element of the heap to the top,
        self.vec.pop()
        self.len -= 1
        father_id = 0 
        while 2 * father_id + 2 <= self.len: # then perform the sift-down operation
            left_child_id = 2 * father_id + 1
            right_child_id = 2 * father_id + 2
            select_id = left_child_id
            if right_child_id < self.len and self.vec[right_child_id] < self.vec[left_child_id]:
                select_id = right_child_id
            if self.vec[father_id] <= self.vec[select_id]: # swap with the smaller child
                break
            self.vec[select_id], self.vec[father_id] = self.vec[father_id], self.vec[select_id]
            father_id = select_id
        return top_val

    def heapify(self, vec):
        for dat in vec:
            self.push(dat)
        
        
```

```python

vec = [7,2,4,6,8,2,5,4]
print("Input: ", vec)
h = MinHeap(vec)
for _ in vec:
    print(h.pop())
    
vec = []
print("Input: ", vec)
h = MinHeap(vec)
for _ in vec:
    print(h.pop())
```

# Complexity

From the implementation above, we can see

- Insertion: time complexity O(log n), since it may sift up at most log n times
- Deletion: time complexity O(log n), since it may sift down at most log n times
- Build heap: time complexity O(nlog n), since each element requires one insertion.
- top: time complexity O(1)

Space complexity O(n)

            - It's easy to see that the smallest pair overall is nums1[0], nums2[0]
            - So who is the second smallest? It could be nums1[0], nums2[1], or it could be nums1[1], nums2[0]. This is essentially comparing the best combination involving nums1[1] against the best combination involving nums1[0].
            - So who is the third smallest? It could be the best combination involving nums1[1] compared against the best combination involving nums1[0]; or it could be the best combination involving nums1[2] compared against the best combination involving nums1[1]. Which one it actually is depends on the result of the previous comparison.
            - And so on — once k numbers have been popped, we have the result.
            This solution has time complexity O(klogk), since we only perform k pops and 2k pushes (k for initialization, k for the comparisons).

- 🌟[(hard) 41 First Missing Positive](https://leetcode.cn/problems/first-missing-positive/)
    - Approach 1: on the first pass, use a heap to record all positive integers encountered; on the second pass, check whether the target is in the heap — if so, pop it and increment the target. Time complexity: the first pass is O(n), the second involves popping which is O(nlogn), so overall O(nlogn); space complexity O(n) since we need to store the heap.
    - Approach 2: if we want constant space complexity, we can only use the original array itself to store the needed information. Suppose the original array were already sorted with no duplicates and all positive integers — what would the smallest missing number be? It would be the first index i where nums[i] != i + 1, i.e. the i-th value isn't i+1. So we rearrange the original array so that, wherever possible, the value i+1 ends up at index i. When marking a position, swap the value at index i with the value at the index it points to (nums[i]), completing the swap for that target index and bringing its value into position i. Throughout this process, the index currently being worked on stays at i. The conditions for stopping the swap are:
        - The current index and the index it points to (nums[i]) are the same — mark this index as done and move on to a new starting index.
        - The current value doesn't satisfy the swap condition: nums[i] is negative, or it's already been marked as done
        - The value at the target index doesn't satisfy the swap condition: it's <= 0, or it's already been swapped
      Each number undergoes at most one swap, so this process is O(n) overall. Finally, do one more pass: the first index that doesn't satisfy the condition, plus one, is the answer. Otherwise, the answer is the total length + 1
- [(hard) 23 Merge k Sorted Lists](https://leetcode.cn/problems/merge-k-sorted-lists/)
    - Approach 1: push everything into a min-heap, then pop everything out in order. The downside: space complexity is very high — O(kn), where k is the number of lists and n is roughly the length of each list
    - Approach 2: optimize the heap storage by only storing the head of each list. When popping, push the next element after the popped head onto the heap. Note: since Python's heap can't compare ListNode objects directly, we can't store ListNode objects directly, but we can store the index of the list that node belongs to.

```python
# 40 
import heapq
class Solution:
    def getLeastNumbers(self, arr: List[int], k: int) -> List[int]:
        res = []
        if k == 0:
            return res
        for val in arr:
            if len(res) < k or -res[0] > val:
                heapq.heappush(res, -val)
            if len(res) > k:
                heapq.heappop(res)
        return [-val for val in res]
```

```python
# 373
import heapq
class Solution:
    def kSmallestPairs(self, nums1: List[int], nums2: List[int], k: int) -> List[List[int]]:
        heap = [] # use a max-heap to track the k smallest sums found so far
        for num1 in nums1[:k]:
            if len(heap) == k and -heap[0][0] <= num1 + nums2[0]: # early stop
                break
            for num2 in nums2[:k]: # nested loop
                if len(heap) < k or -heap[0][0] > num1 + num2: # compare against the current top of the heap
                    heapq.heappush(heap, [-num1 - num2, num1, num2])
                    if len(heap) > k:
                        heapq.heappop(heap)
                else:
                    break
        return [item[1:] for item in heap]
```

```python
# 373 Approach 2, 
# From the solution https://leetcode.cn/problems/find-k-pairs-with-smallest-sums/solution/gong-shui-san-xie-duo-lu-gui-bing-yun-yo-pgw5/
# and https://leetcode.cn/problems/find-k-pairs-with-smallest-sums/solution/tong-ge-lai-shua-ti-la-you-xian-ji-dui-l-fw7y/
class Solution:
    def kSmallestPairs(self, nums1: List[int], nums2: List[int], k: int) -> List[List[int]]:
        flag, ans = (n := len(nums1)) > (m := len(nums2)), []
        if flag:
            n, m, nums1, nums2 = m, n, nums2, nums1
        pq = []
        for i in range(min(n, k)):
            heapq.heappush(pq, (nums1[i] + nums2[0], i, 0))
        while len(ans) < k and pq:
            _, a, b = heapq.heappop(pq)
            ans.append([nums2[b], nums1[a]] if flag else [nums1[a], nums2[b]])
            if b + 1 < m:
                heapq.heappush(pq, (nums1[a] + nums2[b + 1], a, b + 1))
        return ans
```

```python
# 347
from collections import Counter
import heapq
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        cnts = Counter(nums)
        heap = []
        for key, val in cnts.items():
            if len(heap) < k or val > heap[0][0]:
                heapq.heappush(heap, (val, key))
                if len(heap) > k:
                    heapq.heappop(heap)
        return [item[1] for item in heap]
```

```python
# 1705
import heapq
class Solution:
    def eatenApples(self, apples: List[int], days: List[int]) -> int:
        apple_heap = [] # track (expiration day, remaining apple count)
        res = 0
        for i in range(len(apples)):
            apple = apples[i]
            day = days[i] + i
            heapq.heappush(apple_heap, [day, apple]) # push each day's newly received apples onto the heap
            while apple_heap and apple_heap[0][0] <= i: # discard expired apples
                heapq.heappop(apple_heap)
            if apple_heap: # decide to eat one apple
                top_day_apple = heapq.heappop(apple_heap)
                res += 1
                if top_day_apple[1] > 1 and top_day_apple[0] > i + 1: # decide whether to push the leftover apples back onto the heap; if they'll expire tomorrow, just discard them.
                    heapq.heappush(apple_heap, [top_day_apple[0], top_day_apple[1] - 1])
        
        # keep eating apples until they're all eaten or all spoiled
        i += 1
        while True:
            while apple_heap and apple_heap[0][0] <= i:
                heapq.heappop(apple_heap)
            if apple_heap:
                top_day_apple = heapq.heappop(apple_heap)
                res += 1
                if top_day_apple[1] > 1 and top_day_apple[0] > i + 1:
                    heapq.heappush(apple_heap, [top_day_apple[0], top_day_apple[1] - 1])
                i += 1
            else:
                break
        return res
```

```python
# 987
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
import heapq
class Solution:
    def verticalTraversal(self, root: TreeNode) -> List[List[int]]:
        heap = []
        left_nodes = [(root, 0, 0)]
        while left_nodes:
            top, row_id, col_id = left_nodes.pop()
            if top.left:
                left_nodes.append((top.left, row_id + 1, col_id - 1))
            if top.right:
                left_nodes.append((top.right, row_id + 1, col_id + 1))
            heapq.heappush(heap, [col_id, row_id, top.val])
        last_col = None
        res = []
        while heap:
            top_item = heapq.heappop(heap)
            if last_col is None or top_item[0] != last_col:
                last_col = top_item[0]
                res.append([top_item[2]])
            else:
                res[-1].append(top_item[2])
        return res
```

```python
# 871  TLE 
import heapq
class Solution:
    def minRefuelStops(self, target: int, startFuel: int, stations: List[List[int]]) -> int:
        fuel_heap = [[0, startFuel]]
        for mile, gas in stations:
            while fuel_heap and fuel_heap[0][1] < mile: # for each station, remove records that can't reach this station
                heapq.heappop(fuel_heap)
            if not fuel_heap: # if no record can reach this gas station, return -1 directly
                return -1
            refuel_now = []
            for base_time, base_gas in fuel_heap: 
                if base_gas > target:
                    continue
                refuel_now.append([base_time + 1, base_gas + gas]) # record the resulting total fuel and refuel count for refueling at this station
            for cur_time, refuel_gas in refuel_now: # push into the heap
                heapq.heappush(fuel_heap, [cur_time, refuel_gas])
        while fuel_heap and fuel_heap[0][1] < target:
            heapq.heappop(fuel_heap)
        if not fuel_heap:
            return -1
        return heapq.heappop(fuel_heap)[0]
```

```python
# 41 Approach 1: Heap
import heapq
class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        possible_heap = [] # use a heap to track the numbers found
        for num in nums:
            if num <= 0:
                continue
            heapq.heappush(possible_heap, num) 
        aim_int = 1 # the target number
        while possible_heap and aim_int == possible_heap[0]: # if the target number is found, pop it and increment the target
            while possible_heap and aim_int == possible_heap[0]:
                heapq.heappop(possible_heap)
            aim_int += 1
        return aim_int
```

```python
# 41 Approach 2: in-place marking
class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        start_id = 0
        max_val = len(nums) + 1
        for i in range(max_val - 1): # first pass, handle all possible noise values, i.e. stored values exceeding the largest possible positive integer we care about
            if nums[i] >= max_val:
                nums[i] = -1
        while start_id < max_val - 1: # second pass, mark positions that satisfy the condition
            if nums[start_id] == start_id + 1: # the next position to swap with is itself, condition satisfied, move on to processing the next starting position.
                nums[start_id] = max_val
                start_id += 1
            else:
                if nums[start_id] <= 0 or nums[start_id] == max_val or nums[nums[start_id] - 1] == max_val:
                    # decide whether to swap with the next number: skip if the current value is negative, already satisfies the condition, or the next number doesn't need swapping
                    start_id += 1
                else:
                    # swap with the next number
                    next_id = nums[start_id] - 1
                    nums[start_id] = nums[next_id]
                    nums[next_id] = max_val
        for i in range(max_val -  1): # third pass, find the smallest index that doesn't satisfy the condition
            if nums[i] != max_val:
                return i + 1
        return max_val
```

```python
# 23 Approach 1: push everything into a stack
import heapq
class Solution:
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        heap = []
        for sub_list in lists:
            while sub_list:
                heapq.heappush(heap, sub_list.val)
                sub_list = sub_list.next
        head = None
        res = None
        while heap:
            next_node = ListNode(heapq.heappop(heap))
            if head is None:
                head = next_node
                res = head
            else:
                res.next = next_node
                res = res.next
        return head
```

```python
# 23 Approach 2: only push the head value of each list onto the heap
import heapq
class Solution:
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        heap = []
        for i, sub_list in enumerate(lists):
            if sub_list:
                heapq.heappush(heap, (sub_list.val, i)) # Python can't compare ListNode objects, so push the index of the corresponding list onto the heap instead.
                lists[i] = sub_list.next
        head = ListNode(0) # clever use of a dummy node!
        res = head
        while heap:
            top_val, top_id = heapq.heappop(heap)
            if lists[top_id]: # when popping, check whether to push the next element back onto the heap.
                heapq.heappush(heap, (lists[top_id].val, top_id))
                lists[top_id] = lists[top_id].next

            next_node = ListNode(top_val)
            res.next = next_node
            res = res.next
        return head.next
```

```python
# 480
import heapq
class Solution:
    def __init__(self):
        self.left_heap = [] # a maxheap, tracks the smaller half
        self.right_heap = [] # a minheap, tracks the larger half
        self.buffer = {} # records of lazily-deferred pops
        self.left_extra_len = 0 # how many extra elements on the left still need to be popped but haven't been yet, used to compute size balance
        self.right_extra_len = 0 # how many extra elements on the right still need to be popped but haven't been yet, used to compute size balance

    def findMedian(self):
        if len(self.left_heap) - self.left_extra_len == len(self.right_heap) - self.right_extra_len:
            return (-self.left_heap[0] + self.right_heap[0])/2.0
        else:
            return -self.left_heap[0]

    def addNumber(self, num):
        if not self.left_heap:
            heapq.heappush(self.left_heap, -num)
            return
        if num <= -self.left_heap[0]:  # must match the pop condition
            heapq.heappush(self.left_heap, -num)
        else:
            heapq.heappush(self.right_heap, num)
        self.prune()
    
    def prune(self):
        # clean up the heap tops, and maintain the balance between the two sizes
        left_true = False
        right_true = False
        while self.left_extra_len > 0 and self.buffer.get(-self.left_heap[0], 0) > 0:
            self.buffer[-heapq.heappop(self.left_heap)] -= 1
            self.left_extra_len -= 1
        while self.right_extra_len > 0 and self.buffer.get(self.right_heap[0], 0) > 0:
            self.buffer[heapq.heappop(self.right_heap)] -= 1   
            self.right_extra_len -= 1

        # maintain the balance of the left heap and right heap
        while len(self.left_heap) - self.left_extra_len - 1 > len(self.right_heap) - self.right_extra_len:
            to_move = -heapq.heappop(self.left_heap)
            if self.buffer.get(to_move, 0) > 0:
                self.buffer[to_move] -= 1
                self.left_extra_len -= 1
            else:
                heapq.heappush(self.right_heap, to_move)

        while len(self.right_heap) - self.right_extra_len > len(self.left_heap) - self.left_extra_len:
            to_move = heapq.heappop(self.right_heap)
            if self.buffer.get(to_move, 0) > 0:
                self.buffer[to_move] -= 1
                self.right_extra_len -= 1
            else:
                heapq.heappush(self.left_heap, -to_move)
                
        if not self.left_heap or self.buffer.get(-self.left_heap[0], 0) == 0:
            left_true = True
        if not self.right_heap or self.buffer.get(self.right_heap[0], 0) == 0:
            right_true = True
        if left_true and right_true: # only stop once the sizes are balanced and neither heap's top is an element pending cleanup; otherwise keep cleaning.
            return
        self.prune()

    def delNumber(self, num):
        self.buffer[num] = self.buffer.get(num, 0) + 1
        if num <= -self.left_heap[0]: # must match the push condition
            self.left_extra_len += 1
        else:
            self.right_extra_len += 1
        self.prune()

    def medianSlidingWindow(self, nums: List[int], k: int) -> List[float]:
        res = []
        for i in range(k):
            self.addNumber(nums[i])
        res.append(self.findMedian())
        for i in range(k, len(nums)):
            self.delNumber(nums[i-k])
            self.addNumber(nums[i])
            res.append(self.findMedian())
        return res
```

# My Summary

- Two heaps: for the streaming median problem 259, use a max-heap and a min-heap respectively to maintain the smaller half and the larger half; the advanced version, 480, adds lazy popping.
- TopK: should you use a min-heap or a max-heap? It depends on the push condition.
    - If the numbers you want to keep should be as small as possible (smallest top-k), use a max-heap: the current largest sits on top, push if the new number is smaller, and pop once it overflows;
    - If the numbers you want to keep should be as large as possible (largest top-k), use a min-heap: the current smallest sits on top, push if the new number is larger, and pop once it overflows;
    
    
    
Advantages and disadvantages of a heap:
- Advantage: quickly find the "extreme" value
- Disadvantage: the remaining elements are left unordered

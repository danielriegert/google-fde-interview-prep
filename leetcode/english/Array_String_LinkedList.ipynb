# Theory

There are only two ways to store a data structure: linked list (chained storage) and array (sequential storage)
Basic operations on data structures: traversal + access, or more specifically, insert/delete/search/update

- Traversal and access come in two basic forms: linear (iterative) and nonlinear (recursive)
    - Linear: array traversal, a typical iterative structure, for/while
    - Nonlinear: recursive structure. Linked lists can be traversed either linearly or nonlinearly.
        - From linked-list traversal, to binary-tree traversal, to N-ary-tree traversal, to graph traversal (using a visited array to record state)

- Linked list: advantage: insert/delete
- Array: advantage: lookup

## Linked List

A small trick for linked-list **deletion** operations:
Operating on the current node directly often causes memory or pointer issues. There are two small tricks that solve this:
- One is to try to operate on the current node's next node rather than the current node itself,
- The other is to create a dummy node pointing to the current list's head node, so that even if every node in the original list gets deleted, the dummy node still exists, and you can simply return dummy->next.


The main thing to think about is when to perform the operation. Do you act on the next node from the current node? Or do you keep a pre node and act on the current node?
Use iteration or recursion? Recursion requires extra stack space.


### Commonly Used Techniques

- Dummy head node: if the head node might change, or the list might be empty, set up a dummy head node pointing to the real head. This eliminates a lot of separate case analysis. For example, 21 Merge Two Sorted Lists (the list may be empty), 19 Remove the N-th Node From End (the head node might get removed)

- Fast/slow pointers: uses
    - Counting within a list of unknown length, e.g. problem 19
    - Finding a position with a specific property in a list, e.g. problems 141, 160
    - Determining whether a list has a cycle: whether the fast and slow pointers coincide; advanced: finding the start of the cycle -- reset the slow pointer to the head, and the point where they meet again is the cycle's start. 141, 142, 160

- Pre-order/post-order traversal: since a linked list also has a recursive structure, it can be traversed in pre-order or post-order as well

```
void traverse(ListNode head) {
    // pre-order traversal code
    traverse(head.next);
    // post-order traversal code
}
```

### Reverse Linked List

In order of increasing difficulty: reverse the entire list (206), reverse the first n nodes of a list, reverse an arbitrary specified segment of a list (92). Both recursion and iteration work. However, recursively operating on a linked list isn't very efficient -- compared to the iterative solution, although both have O(N) time complexity, the iterative solution has O(1) space complexity, while the recursive solution needs a call stack, giving O(N) space complexity. So for efficiency's sake, go with iteration~

- [(easy)206 reverse linked list](https://leetcode.cn/problems/reverse-linked-list/)
    - Approach 1: recursion. Reverse the next node, then place head at the end of the reversed portion. Watch out for edge cases. Space complexity O(n) -- the space complexity is mainly determined by the recursive call stack, which is at most n levels deep.
    - Approach 2: iteration. My implementation keeps creating new nodes. Space complexity O(n)
    - Approach 2 (variant): iteration, without creating new nodes -- each time, first record the current element's next element, then set the current element's next to the old head, then set the current element as the new head. Space complexity O(1)
    
    The difference in mindset between recursion and iteration:
    - Recursion: the question to ask is: if the rest of the list after me has already been reversed, what do I do? Place myself at the end of the reversed list, and return the new list.
    - Iteration: the question to ask is: if the list before me has already been reversed, what do I do? Next time I should start from my next node; I should connect myself to the head of the reversed list and become the new head.
    
- [(medium)92 Reverse Linked List II](https://leetcode.cn/problems/reverse-linked-list-ii)
    - Approach 1: recursion. First find the position right before where reversal should start, since that position's next value needs to change. Then reverse the first n nodes of the list. Finally, connect the reversed list after that earlier position.
    - Approach 2: iteration. Find the position pre right before where reversal should start. Within the range to be reversed, maintain a tail element whose next points to the current element being reversed; update tail.next to the element-to-reverse's next, update the element-to-reverse's next to pre.next, and update pre.next to the element-to-reverse.
    

- [(hard)25 Reverse Nodes in K group](https://leetcode.cn/problems/reverse-nodes-in-k-group)
    - Approach 1: recursion overall, iteration locally. Once the first k nodes are reversed, the rest can be solved recursively. Reversing the first k nodes can be done either recursively or iteratively; here iteration is used.
    - Approach 2: recursion overall, recursion locally. Reversing the first k nodes can also be done recursively

````python
# 234 O(1) space solution
# Runtime: 672 ms, beats 65.04% of Python3 submissions
# Memory: 40.5 MB, beats 73.61% of Python3 submissions
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def findMiddle(self, head):
        faster = head
        slower = head
        while True:
            if not faster or not faster.next:
                break
            faster = faster.next.next
            slower = slower.next
        return slower
    
    def reverse(self, head):
        tail = head
        old_head = tail
        while tail.next:
            new_head = tail.next
            tail.next  = tail.next.next
            new_head.next = old_head
            old_head = new_head
        return old_head

    def isPalindrome(self, head: ListNode) -> bool:
        middle = self.findMiddle(head)
        second_head = self.reverse(middle)
        p1 = head
        p2 = second_head
        while p2:
            if p1.val != p2.val:
                return False
            p1 = p1.next
            p2 = p2.next
        return True
````

# Lessons Learned

    Cleverly construct a dummy head node. This makes traversal logic more uniform.
    Use recursion flexibly. Constructing the right recursive condition lets recursion solve problems elegantly. But note that some problems can't use recursion, since too deep a recursion depth causes timeouts and stack overflows.
    Reversing a range within a linked list. Problem 92.
    Finding the middle node of a linked list. Problem 876. Finding the n-th node from the end of a linked list. Problem 19. Both can be solved with a single traversal.
    Merging K sorted linked lists. Problems 21 and 23.
    Partitioning a linked list. Problems 86 and 328.
    Sorting a linked list with required time complexity O(n * log n) and space complexity O(1). There's only one approach: merge sort, top-down merging. Problem 148.
    Determine whether a linked list has a cycle; if so, output the index of the cycle's entry point. Determine whether two linked lists intersect; if so, output the intersection point. Problems 141, 142, and 160.

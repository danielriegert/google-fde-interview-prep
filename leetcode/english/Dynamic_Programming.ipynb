# Knowledge Points

The overall guiding idea of dynamic programming is **to reduce useless or duplicate computation**.
The essence of this method is to obtain the result of a larger-scale problem from the results of several smaller-scale sub-problems.

- [X] When can dynamic programming be used to solve a problem?

Usually we should start by analyzing whether the problem has **no aftereffect** (memorylessness). If, for a given state, we only need to care about the value of the state and not how the state was transitioned to, then this is a memoryless problem, and DP can be considered.

Another more practical trick: we can also guess whether DP is applicable based on the **data range**.

Because DP is a recurrence process,
- If the data range is 10^5-10^6, consider whether a 1D DP can solve it
- If the data range is 10^2-10^3, consider whether a 2D DP can solve it


Categorizing the DP problems you've done by these types, and **analyzing the similarities in how states are represented and how the state transition equations are constructed** within each type, will help you learn dynamic programming faster.

## DP Template

Dynamic programming and recursion (with state) are essentially the same thing — both search the entire state space for a way to avoid "brute-force enumeration".
- "Top-down" solving via "recursion",
- Dynamic programming: "bottom-up" solving via "recurrence".

Three elements of dynamic programming:
- Has "optimal substructure": the optimal result of a larger-scale problem can be derived from the optimal results of its sub-problems. To satisfy "optimal substructure", the sub-problems must be independent of each other.
- Has "overlapping sub-problems": use a "memo" or a "DP table" to optimize the brute-force process and avoid unnecessary computation.
- A correct state transition equation: this is a case analysis of the last step.

### Optimal Substructure


Defines the relationship between a sub-problem and the original problem.

If a problem can be decomposed into multiple sub-problems, then recursively find the optimal solution of each sub-problem, and finally combine the optimal solutions of each sub-problem through some mathematical method to obtain the final result. In summary, the optimal solution of a problem is determined by the optimal solutions of its sub-problems.

- [X] What does it mean for sub-problems to be independent of each other?
To satisfy optimal substructure, sub-problems must be independent of each other. If the solution to the original problem could arise from a combination across two sub-problems, then clearly the sub-problems are not independent. For example: knowing the maximum score difference within each class, find the maximum score difference across the whole school. This does not satisfy optimal substructure, because the maximum score difference for the whole school might occur between two different classes.

- [X] What do we do when we encounter a case where optimal substructure fails like this?
The strategy is: reformulate the problem, **transform it into an equivalent problem**. For example, the whole-school score-difference problem. Reformulating the problem — isn't "maximum score difference" equivalent to "the difference between the highest score and the lowest score"? So isn't it really about finding the highest and lowest scores?

**Being able to combine the solutions of sub-problems to obtain the solution of the original problem is the key to whether dynamic programming is feasible.**


Optimal substructure is not a property unique to dynamic programming; most problems that ask for an optimal value have this property. But conversely, since optimal substructure is a necessary condition for a DP problem, it must be one asking you to find an optimal value — so whenever you encounter one of those annoying "find the optimal value" problems in the future, think dynamic programming — that's the pattern.


### Overlapping Sub-problems

Overlapping sub-problems defines the relationship between sub-problems themselves.

Overlapping sub-problems mean that direct recursion involves a lot of repeated computation, which leads to the idea of **memoized recursion (essentially building a memo)**: if the range of sub-problems can be determined in advance, we can build a table to store the answers to sub-problems.

Key points for understanding optimal substructure and overlapping sub-problems in DP algorithms

1. Prove that the solution to the problem includes a choice, and after making the choice, one or more sub-problems remain.

2. Design a recursive description of the sub-problems.

3. Prove that the optimal solution to the original problem is composed of the optimal solutions of all the sub-problems.

4. Prove that the sub-problems overlap (this step is not required for the correctness of dynamic programming, but if the sub-problems do not overlap, the efficiency is the same as plain recursion)



### State Transition Equation

**The state transition equation is a case analysis of the last step**.

In problem solving, the state transition equation is generally used to describe this combination.

For example, suppose the solution to the original problem is $f(n)$, where $f(n)$ is also called the state. The state transition equation $f(n)=f(n−1)+f(n−2)$ describes a combination relationship between the original problem and its sub-problems.

If you guess a state definition and then find you cannot write a state transition equation that covers all cases (without missing any), it's likely that the state definition was guessed wrong — switch approaches quickly instead of struggling with the state transition equation.

Whether our state transition needs to be "without missing any" or "without missing any and without duplicates" depends on the problem itself:
- If we're finding an optimal value, we only need to ensure "no cases are missed", because duplicates don't affect the result.
- If we're counting the number of ways, we need to ensure "no duplicates and no cases missed".



Analyze the problem step by step through the following 4 elements:

1. What is the state: determine the dp array (dp table) and the meaning of its indices
2. What is the state transition equation: determine the recurrence formula
3. What are the initial values of the states: in some cases the recurrence formula determines how the dp array should be initialized!
4. What is the final answer the problem is asking for


Once each step has been analyzed, the entire DP problem is essentially solved.

### Traversal Direction of the dp Array
Take a 2D dp array as an example — sometimes we traverse forward, sometimes backward, and sometimes even diagonally.

Just keep these two points in mind:

1. During traversal, the states required must have already been computed.

2. After traversal ends, the position storing the result must have already been computed.

## Complexity Analysis

For DP complexity/computation analysis, the complexity/computation amount equals the number of states.

Therefore, 1D DP complexity is usually linear, while 2D DP complexity is usually quadratic.

## Categories

### Knapsack Problems

Knapsack problems are a very classic category within "dynamic programming". Knapsack problems are essentially an **NP-complete problem** within combinatorial optimization.

- "NP-complete problems" can be simply understood as problems that "cannot be solved directly". They can only be solved via "brute force" + "verification".

"Knapsack problems" generally correspond to this type of problem:

Broadly speaking, given **values and costs**, along with **restricted decision rules**, how do we **maximize value** under these conditions.

For example: a "knapsack" refers to being given the **value and volume** of items (corresponding to "given values and costs"), and under a **fixed capacity** (corresponding to "restricted decision rules"), how to **maximize the total value of the selected items**.

- 0/1 Knapsack: each item can be selected at most once. The state transition equation is
$
dp(i,j) = max(dp(i-1,j), dp(i-1, j-vi)+wi)
$

- Unbounded (complete) Knapsack: each item can be selected an unlimited number of times. There are n kinds of items and a knapsack of capacity C, and each item has unlimited supply. The volume of the i-th item is vi, and its value is wi. Solve for which items to put in the knapsack so the total cost does not exceed the knapsack's capacity, while maximizing total value. The state transition equation is
$
dp(i,j) = max(dp(i-1,j), dp(i, j-vi)+wi)
$

The difference is in the second term: unbounded knapsack still looks at i, while 0/1 knapsack looks at i-1. Both can be compressed to 1D space,
- 0/1 knapsack depends on the previous layer i-1's value at j-vi, so you must traverse from large to small, to ensure that i-1's j-vi hasn't already been overwritten by row i.
- Unbounded knapsack depends on the current layer i's value at j-vi, so you must traverse from small to large, to ensure that i's j-vi has already been updated to row i's value.


- Bounded (multiple) knapsack: each item has a limited quantity. Optimization approaches:
    - Flattening/expansion: splits the quantity si of an item into multiple items of quantity 1, converting it into a "0/1 knapsack".
        - The most direct way to split: split into si items each with volume vi and value wi. The downside of this approach is that if the original solution took one of this item, under the new split it corresponds to si possible cases — i.e., separately taking the 1st, 2nd, ..., si-th. This increases the enumeration complexity. Overall complexity is $O(\sum_i s_i C)$
        - "Binary splitting": can greatly reduce duplication. This splits si using binary numbers. For example, if si is 10, first take out 1, 10-1=9, then take out 2, 9-2=7, then take out 4, 7-4=3, stop. So 10 is split into 1, 2, 4, 3 — 4 items, whose volumes/values are 1x, 2x, 4x, 3x the original respectively, each with quantity 1. This way, if the original solution took 1 item, it corresponds to taking item 1; if it took 2, it corresponds to taking item 2; if it took 3, it corresponds to taking items 1 and 2, or item 3; if it took 4, it corresponds to taking item 4, or items 1 and 3; if it took 5, it corresponds to taking items 1 and 4, or 2 and 3; if it took 6, items 2 and 4; if it took 7, items 1, 2, and 4, or 3 and 4; if it took 8, items 1, 3, and 4; if it took 9, items 2, 3, and 4; if it took 10, all of 1, 2, 3, and 4. This greatly reduces duplicate cases. Overall complexity is $O(\sum_i \log s_i C)$
    - Monotonic queue optimization: used to resolve the innermost "decision" loop in bounded knapsack. Uses the sliding-window-maximum idea. If you want to always take the value from the front of the queue for updates, you need to maintain "monotonicity of queue elements" and "a specific window size". [Details reference](https://mp.weixin.qq.com/s?__biz=MzU4NDE3MTEyMA==&mid=2247486963&idx=1&sn=51d34f0f841122ed9be2335a402041e8&chksm=fd9ca6eccaeb2ffa1abe413177be376799b427b092bfb73c13e7b77e171b460f4c24b3b7d3bc&cur_album_id=1751702161341628417&scene=189#wechat_redirect)

The "1D space optimization" approach has now been covered for all three traditional types of knapsack problems.

We find that only the "unbounded knapsack"'s "1D space optimization" is a mathematically meaningful optimization (it can effectively reduce the algorithm's time complexity).

The "1D space optimization" for "0/1 knapsack" and "bounded knapsack" is actually just a "rolling" operation performed on top of the naive 2D solution (leveraging the dependency relationships between states, combined with the traversal order, so that space no longer needed for the transition can be reused).

#### Summary

- 0/1 Knapsack: emphasizes that each item "can only be selected once". Performing "1D space optimization" on it does not reduce time complexity; when doing "1D space optimization", the capacity dimension must be traversed "from large to small".

- Unbounded Knapsack: emphasizes that each item "can be selected an unlimited number of times". Performing "1D space optimization" on it is mathematically meaningful and can reduce time complexity from $O(NC^2)$
to $O(NC)$, when doing "1D space optimization", the capacity dimension must be traversed "from small to large".

- Bounded Knapsack: emphasizes that each item "can only be selected a limited number of times". Neither "1D space optimization" nor "plain flattening" reduces its time complexity — additional optimization techniques must be applied: binary optimization or monotonic queue optimization.

- Mixed knapsack problem: the knapsack contains all three of the above item types. The outer loop of a knapsack problem can iterate over items!
    1. First convert any "bounded knapsack" items into "0/1 knapsack" items via the "binary optimization" idea.
    2. Then, depending on whether an item belongs to "0/1 knapsack" or "unbounded knapsack", decide whether to compute the capacity dimension "from large to small" or "from small to large". In other words, choose a different transition method depending on the item's type.

- Grouped Knapsack: group multiple items together, and require that at most one item be taken from each group. Method: the outermost loop iterates over groups, and within each group, iterate over which item to take.

At the same time, the first three types of knapsack problems all have two kinds of state definitions: **at most** and **exactly**.

These two state definitions only differ in "initialization".

As for how to initialize, the key is to grasp which states are valid:

- For the "at most" state definition: $f[0][x]$ is always a valid value of 0 — meaning that without considering any items, the maximum value obtainable with knapsack capacity "at most $x$" is 0.

- For the "exactly" state definition: among $f[0][x]$, only $f[0][0]$ is a valid value of $0$; all other values are "invalid". This represents that without considering any items, only when the knapsack capacity is "exactly 0" is the maximum obtainable value 0; other capacities that are "exactly nonzero" cannot achieve a valid value (because no items are considered).

````python
# 887 DP + binary search + space optimization
# time 9772 ms
# space 15.4MB
class Solution:
    def findMin(self, diff, egg): # binary search
        l = 1
        r = diff
        while l < r:
            m = (l + r) // 2 + 1
            if self.memo[1-egg][m-1]>=self.memo[egg][diff - m]:
                r = m - 1
            else:
                l = m
        smaller_i = l
        return min(self.memo[egg][diff-smaller_i], self.memo[1-egg][smaller_i])
        
    def superEggDrop(self, k: int, n: int) -> int:
        self.memo = [[i for i in range(n+1)] for _ in range(2)]
        for i in range(2):
            self.memo[i][0] = 0
        for egg in range(1, k): # the egg number is egg + 1
            for diff in range(1, n + 1):
                cur_min = self.findMin(diff, egg % 2)
                self.memo[egg % 2][diff] = 1 + cur_min
        return self.memo[(k - 1) % 2][n]
````

    - Approach 1: DFS and dynamic programming are actually quite similar — both break a big problem into sub-problems, then solve the big problem by solving the sub-problems. The DFS idea is intuitive. I split s1 in every possible way into x+y, then recursively check whether either of the following two cases holds:
        - x matches the corresponding-length prefix segment of s2, and y matches the corresponding-length suffix segment of s2
        - x matches the corresponding-length suffix segment of s2, and y matches the corresponding-length prefix segment of s2
         If either holds, this recursive call's s1 and s2 are considered a match.
         Points to note: each matched/unmatched state is jointly marked by the segment's starting position in s1 and its starting position in s2, not just by s1. Characteristics of interval problems:
- Since we need to enumerate both the start and end of the interval, there are n^2 possibilities in total, i.e. space complexity is n^2.
- Generally start enumerating from interval length 0 or 1, and iterate based on interval length
- State representation: interval start coordinate + the required value within the interval
- State transition equation: the value of a large interval is composed from smaller sub-intervals — enumerate the possibilities for the sub-intervals


Template
1. initialize the subpart with the length of 0 or 1
2. iterate on the subpart length
    3. iterate on the subpart start_id: get the value of matrix[start_id][start_id + len] based on the sub-subparts within it.

````python
# 375 dp
class Solution:
    def getMoneyAmount(self, n: int) -> int:
        allguess = [[-1 for i in range(1+n)] for _ in range(1+n)]
        # 1. initialize the case where interval length is 0
        for i in range(1+n):
            allguess[i][i] = 0
        # 2. iterate over the interval length
        for diff in range(1, n):
            # iterate over the interval start for a fixed interval length
            for start in range(1, 1 + n - diff):
                # update the state of the target interval based on the states of its sub-intervals
                min_guess = min(start + allguess[start + 1][start + diff], allguess[start][start + diff - 1] + start + diff)
                for k in range(start+1, start + diff):
                    min_guess = min(min_guess, k + max(allguess[start][k - 1], allguess[k+1][start + diff]))
                allguess[start][start + diff] = min_guess
        return allguess[1][n]
````

For each current i, the array holds the information for all amounts at that i, i.e. what was originally status[i][0], status[i][1], ..., status[i][amountmax] in 2D. Now, when we want the information for the next i+1, can our array satisfy this? f(i+1, cur_amount) = f(i, cur_amount) + f(i+1, cur_amount-coins[i]) — clearly yes, because f(i, cur_amount) has been saved, and the initial state of f(i+1, cur_amount-coins[i]) can be obtained via initialization, with other states updated iteratively.
       In summary: because in the state transition equation, f(i,amount) and f(i-1,amount) are directly adjacent, they can be directly overwritten by an update without impact; but f(i,amount) and f(i,amount-coins[i]) are not directly adjacent, so overwriting would affect the result. So the outer loop iterates over i (once finished, it will overwrite the array), and the inner loop iterates over amount (updating the 1D array).

What do we do? The clever trick is to use an extra variable to store the previous value of dp[j-1], so when dp[i+1][j-1] is needed, we use the stored value to compute it. # Summary

- [X] When to use recursion, and when to use DP?

At a higher level, both recursion and DP search a graph of the state space to find a path to the target node. Each state can be viewed as a node in the graph. Recursion generally walks backward step by step from the end state, and doesn't necessarily visit every state. DP, on the other hand, picks a starting state and walks forward step by step to the end state, visiting every possible state along the way.

In summary,
- If both the end state and the start state are clearly defined, DP and recursion have similar complexity, but DP uses iteration, which is much simpler to implement, and might allow space optimization. In this case, DP is the better choice. Example problem 740
- If the end state is clear but the start state is not, DP needs to traverse all states, while recursion goes from back to front and doesn't necessarily need to visit all states — in this case, recursion is often the better choice. Example problem 87.

Detailed analysis based on the example problems:

Problem 87, recursion + memoization takes 68ms, DP takes 996ms. DP's number of visits is significantly more than recursion's. Because DP's starting states cover all possible states, DP has to walk through essentially every state to reach the end; recursion, on the other hand, is goal-directed, working backward from the end and only searching part of the state space. In this case, recursion is better.

Problem 740, recursion + memoization takes 336ms, DP takes 44ms. In this problem, the two methods actually search the same state tree, but because recursive calls require extra space allocation, overall it's slower than iteration — so from an implementation standpoint, DP is the better choice here.

- [X] When to use an array to record states, and when to use a dictionary?
    - If all states must be recorded: use an array. DP generally needs to record all states, and there's also the space-optimization trick
    - If only some states need to be recorded, use a dictionary. This saves unnecessary overhead.

   Common ground between the two: both support O(1) reads and writes.


- [X] Space compression trick: when can it be used? Along which axis should compression happen?
    - If the current state's update can be derived from adjacent states, you can try space compression. Traverse along the adjacent axis, updating the state of the other dimension.

- [X] Why does example problem 332 give correct results when compressed along either direction? What about example problem 518?
    - 332 is a coincidence — the min function happens to be "transitive" in this sense. The truly correct approach should be to iterate along the coin dimension, updating the amount information.
    - 518's sum function is not transitive in this way. You can only iterate along the coin dimension. If you iterate along the amount dimension, then for each amount, when you reach i, the state you update actually represents the count of combinations after adding coin[i] to amount — taken together, what you get is a permutation count, not a combination count.

## Algorithms on Graphs

### BFS Template

1. Arbitrarily choose a point as the starting point, and count it into the current layer
2. Mark all points in the current layer as visited. For each point in the current layer, look at its neighbors. Add any not-yet-visited neighbors to the next layer.
3. If the next layer isn't empty, set it as the current layer and repeat step 1.
4. If the next layer is empty, depending on the problem, either pick a new starting point or end the search.


Bidirectional BFS template


Dijkstra's algorithm can be viewed as breadth-first search optimized with a greedy strategy


### DFS Template

1. Arbitrarily choose a point as the starting point
2. If this starting point has no unvisited neighbors, end the DFS
3. Among this starting point's unvisited neighbors, pick one and repeat step 1 until done; then move on to the next unvisited neighbor, until there are none left.

### Topological Sort

This is actually a special variant of BFS. For each step of BFS, the corresponding variant in topological sort is:
1. Choose all points with in-degree 0 as starting points, and count them into the current layer
2. For each point in the current layer, look at its neighbors (the nodes it points to), and decrement each neighbor's in-degree by one. If a neighbor's in-degree drops to 0, add it to the next layer.
3. If the next layer isn't empty, set it as the current layer and repeat step 1.
4. If the next layer is empty: if there's still a point with nonzero in-degree, it's a failure; otherwise, it's a success -- return according to the problem's requirements.


- [X] Under what circumstances would the next layer be empty while some point still has nonzero in-degree?
If and only if there's a cycle

### Eulerian Path

The "one-stroke drawing" problem: traverse every edge exactly once
Eulerian circuit: the start and end points coincide, forming a cycle
Eulerian path: the start and end points don't coincide

A tree that includes all n points of the graph and has only n-1 edges

```python
# 785 multiple rounds of BFS
class Solution:
    def __init__(self):
        self.color = []

    def getUncolored(self):
        res = []
        for i, c in enumerate(self.color):
            if c == 0:
                res.append(i)
        return res
    
    def coloring(self, start_node, color):
        next2color = set([start_node])
        while next2color:
            cur2color = next2color
            next2color = set()
            for node in cur2color:
                self.color[node] = color
                for neigh in self.graph[node]:
                    if self.color[neigh] == 0:
                        next2color.add(neigh)
                    elif self.color[neigh] == color:
                        return False
            color = 3 - color
        return True

    def isBipartite(self, graph: List[List[int]]) -> bool:
        self.color = [0] * len(graph)
        self.graph = graph
        while True:
            uncolored_nodes = self.getUncolored() # this kicks off multiple rounds of coloring
            if not uncolored_nodes:
                return True
            start_node = uncolored_nodes[0]
            if not self.coloring(start_node, 1): # whether this round of coloring succeeded
                return False
        return False
```

- [(hard)127Word Ladder](https://leetcode.cn/problems/word-ladder/)
    - Think about the following questions
        - [X] How does word length affect the choice of algorithm? Call it `L` -- it affects time complexity linearly
        - [X] How does the number of possible substitutions per position affect the choice of algorithm? Call it `C`
        - [X] How does the dictionary size affect the choice of algorithm? Call it `M`
        - [X] When do you determine a node's children? There are two ways: pre-build the graph, or find children on the fly during traversal. What's the difference?
            - Pre-building the graph: build a hashmap where the key is each word and the value is the other words that differ from it by one letter. The time complexity is linear in dictionary size `M`. Space complexity: at most O(`M^2`) extra space is needed to store the adjacency relationships.
            - Finding children on the fly: for each word visited during traversal, find its children. The time complexity is linear in the number of traversal steps `R`. The number of traversal steps is always at most `M`.
            - Overall, choose the latter
        - [X] How do you determine a node's children? There are also two ways.
            - One is to iterate over every word in the dictionary and check whether it differs from the target word by one letter, giving time complexity O(`ML`). There are `M` words to iterate over, and each check examines `L` positions.
            - The other is to enumerate all words that differ from this word by one letter, and check whether each is in the dictionary, giving a fixed time complexity of O(`LC`). There are `LC` words to enumerate, and each check is O(1).
            - Based on this analysis, if the dictionary size `M` is smaller than the number of candidates per position `C`, choose iterate-then-check; otherwise choose enumerate-then-check.
        
    - Let's think initially: what approaches can solve this problem?
        - DFS: not suitable -- this is a shortest-path problem, and DFS could waste a lot of effort.
            1. For the current root, find all unvisited children,
                - if the target node is among the children, finish and return
                - otherwise, mark all candidate children as visited
            2. Iterate over all candidate children and recurse into DFS. Does the order of this step matter?
            3. If the current root has no unvisited children, return failure.
            Time complexity: each node might be visited many times, so the time complexity is fairly high -- each node could be visited up to M times, giving up to O(`M^2`) total traversal steps. That's an extra factor of O(`M`) compared to BFS. Space complexity: the call stack could be up to O(`M`) deep.
            
        - BFS：
            1. Iterate over every point in the current layer (at most O(M) iterations)
            2. Find that point's unvisited children -- if this step finds them on the fly during traversal,
                - if that child is the target, finish
                - otherwise, mark the child as visited and add it to the candidates
            3. Treat all candidates as the current layer, and repeat step 1
            - Time complexity: every point is visited at most once; for each point visited, finding its unvisited children costs O(`ML`) or O(`LC`). Overall, the complexity is O(`M^2L`) or O(`MLC`); since `C` can be treated as a constant, the time complexity is O(`ML`).
            - Space complexity: the visited record is O(M); the set of feasible children at each step is at most O(M); everything else can be treated as constant. Overall it's O(M).
            
        - Bidirectional BFS
            1. Run BFS from two sets separately, always expanding the smaller set.
            2. If the two sets intersect, the traversal is done
            - Time complexity: theoretically the same as BFS, but in practice it can be reduced a lot, since it avoids a great deal of unnecessary exploration.

```python
# 127 BFS  
# time 620ms beat 26%; 
# space 16.2MB beat 43%
from collections import deque, defaultdict
class Solution:
    def __init__(self):
        self.target = None
        self.word2children = defaultdict(list)

    def diff_one(self, node, other_node):
        return sum([node[i] != other_node[i] for i in range(len(node))]) == 1
    
    def buildGraph(self, beginWord, wordList):
        all_word = [beginWord] + wordList
        for i, node in enumerate(all_word):
            for other_node in wordList[i: ]:
                if self.diff_one(node, other_node):
                    self.word2children[node].append(other_node)
                    self.word2children[other_node].append(node)

    def findChildren(self, node):
        # find all unvisited children
        # implement two ways to findChildren: 
        # one way is to build a graph in advance and query directly 
        # the other way is to iterate all the possible candidates and find the valid ones.
        # choose according to the dict size with candidate variations size.
        
        children = []
        if self.word_len < 26:
            # build a graph in advance and query directly 
            all_children = self.word2children[node]
            for child in all_children:
                if not self.visited[child]:
                    children.append(child)
        else:
            # iterate all the possible candidates and find the valid ones
            for i in range(len(node)):
                for c in range(26):
                    if chr(ord('a') + c) == node[i]:
                        continue
                    var_node = node[:i] + chr(ord('a') + c) + node[i+1:]
                    if var_node in self.wordList and not self.visited[var_node]:
                        children.append(var_node)
        return children

    def BFS(self, begin_node):
        next_layer = [begin_node]
        path_len = 1
        while next_layer:
            cur_layer = next_layer
            next_layer = []
            for node in cur_layer:
                if self.diff_one(node, self.target):
                    return path_len + 1
                next_layer.extend(self.findChildren(node)) # find all unvisited children
            path_len += 1
        return 0

    def ladderLength(self, beginWord: str, endWord: str, wordList: List[str]) -> int:
        if endWord not in wordList:
            return 0
        self.target = endWord
        self.wordList = set(wordList)
        self.visited = {word: False for word in self.wordList}
        self.visited[beginWord] = True
        self.word_len = len(wordList)
        if self.word_len < 26:
            self.buildGraph(beginWord, wordList)
        return self.BFS(beginWord)
```

```python
# 127 bidirectional BFS
# time 96ms beat 91.76%; 
# space 16.3MB beat 38.43%

from collections import deque, defaultdict
class Solution:
    def __init__(self):
        self.target = None
        self.word2children = defaultdict(list)

    def diff_one(self, node, other_node):
        return sum([node[i] != other_node[i] for i in range(len(node))]) == 1
    
    def buildGraph(self, beginWord, wordList):
        all_word = [beginWord] + wordList
        for i, node in enumerate(all_word):
            for other_node in wordList[i: ]:
                if self.diff_one(node, other_node):
                    self.word2children[node].append(other_node)
                    self.word2children[other_node].append(node)

    def findChildren(self, node, layer_id):
        children = []
        if self.word_len < 26:
            all_children = self.word2children[node]
            for child in all_children:
                if not self.visited[layer_id][child]:
                    children.append(child)
                    self.visited[layer_id][child] = True
        else:
            for i in range(len(node)):
                for c in range(26):
                    if chr(ord('a') + c) == node[i]:
                        continue
                    var_node = node[:i] + chr(ord('a') + c) + node[i+1:]
                    if var_node in self.wordList and not self.visited[layer_id][var_node]:
                        children.append(var_node)
                        self.visited[layer_id][var_node] = True
        return children

    def BiBFS(self, begin_node, end_node):
        self.next_layers = {} # records for both directions
        self.next_layers[0] = [begin_node]
        self.next_layers[1] = [end_node]
        cur_layer_id = 0 # each time, expand the direction with the smaller frontier
        path_len = 2
        while self.next_layers[cur_layer_id]:
            cur_layer = self.next_layers[cur_layer_id]
            self.next_layers[cur_layer_id] = []
            for node in cur_layer:
                children = self.findChildren(node, cur_layer_id) # first find all unvisited points in this direction
                for child in children:
                    if self.visited[1-cur_layer_id][child]: # then check whether it was visited from the other direction
                        return path_len
                self.next_layers[cur_layer_id].extend(children)
            path_len += 1
            if len(self.next_layers[cur_layer_id]) > len(self.next_layers[1-cur_layer_id]):
                cur_layer_id = 1 - cur_layer_id # each time, expand the direction with fewer candidates
        return 0

    def ladderLength(self, beginWord: str, endWord: str, wordList: List[str]) -> int:
        if endWord not in wordList:
            return 0
        self.target = endWord
        self.wordList = set(wordList)
        self.visited = defaultdict(dict)
        for word in self.wordList:
            self.visited[0][word] = False # records for both directions
            self.visited[1][word] = False
        self.visited[0][beginWord] = True
        self.visited[1][endWord] = True
        self.word_len = len(wordList)
        if self.word_len < 26:
            self.buildGraph(beginWord, wordList)
        return self.BiBFS(beginWord, endWord)
```

- [(medium)1631 Path With Minimum Effort](https://leetcode.cn/problems/path-with-minimum-effort/). This problem is a variant of shortest-path. In general, BFS is worth trying for shortest-path problems. When the target is known in advance, bidirectional BFS is also worth trying.
    - Approach 1: BFS
    - Approach 2: Dijkstra. Think about the following questions:
        - [X] Under what conditions can Dijkstra's algorithm be used? When can't it? Can it be used for a problem like 127? Dijkstra's algorithm has an advantage when the candidate points have a notion of "better or worse" among them. In 127, the candidate nodes have no such distinction, so Dijkstra's doesn't bring any advantage there, so it's not used.
        - Complexity analysis: in Dijkstra's algorithm, each edge is selected at most once (complexity O(m), assuming m edges), and selecting the currently best edge each time costs O(log m). Overall the time complexity is O(m log m), and the space complexity is O(m), used to store the candidate edges. In this problem, an m*n grid has (m-1)*n + (n-1)*m = O(mn) edges. So the time complexity is O(mn log(mn)) and the space complexity is O(mn).
    - [X] Approach 3: union-find. The union-find approach checks when the start and end become connected, always connecting nodes via the edge with the least effort. The result is found once start and end are connected. The premise here is that we know the start and end can always be connected -- the key is finding the least-effort path. Union-find needs to enumerate all edges (O(mn)), and each time extract the least-effort one (O(log mn) if using a heap). Overall the time complexity is O(mn log(mn)), and the space complexity is O(mn).
    - [X] Approach 4: binary search. Use binary search to find the extremal value -- the idea is guess + check. This is worth trying whenever there's an efficient check. Can we design an efficient check here? Guess a minimum value; the problem becomes: can we get from start to end without exceeding this effort? That's just a regular BFS problem, visiting each point at most once, with complexity O(mn). If the upper bound on effort is C, the binary search costs O(log C), so overall the time complexity is O(mn log C) and the space complexity is O(mn).
    - [X] Approach 5: can bidirectional BFS be used for this problem? Not optimal, perhaps, but is it possible? In theory yes -- traverse from the start and from the end. But unlike the typical bidirectional BFS scenario, the first point where the two searches meet isn't necessarily on the shortest path. So more traversal is needed -- at minimum, every point needs to be reached from both directions before the global optimum can be determined. So the traversal-count savings that bidirectional BFS usually provides doesn't materialize here. Why not? Because in this problem, the "points to be traversed" have a notion of "better or worse" -- they're weighted. Choosing the currently-best point to expand is itself a significant, arguably the most significant, source of savings.
    **Which traversal method to choose comes down to: in the current context, which method is the most efficient while still reaching the goal.**
    
   Approaches 2 and 3 have similar time complexity, but 2 generally has a memory advantage. Approach 4's time complexity is sensitive to the values in heights. If the maximum value in heights is greater than m*n, it's safer to go with 2 or 3. Of course, if the range of heights is much smaller than m*n, binary search has a clear advantage. Which algorithm to pick really depends on the specifics.

```python
# 1631 BFS
# time 3136ms
# space 16.2MB
class Solution:
    def findNext(self, i, j, heights):
        nexts = []
        for next_i, next_j in [[i-1, j], [i+1, j], [i, j - 1], [i, j + 1]]:
            if next_i < 0 or next_i >= self.m or next_j < 0 or next_j >= self.n:
                continue
            cur_effort = max(self.efforts[i][j], abs(heights[next_i][next_j] - heights[i][j])) # check whether to continue visiting
            if self.efforts[next_i][next_j] == -1 or self.efforts[next_i][next_j] > cur_effort:
                self.efforts[next_i][next_j] = cur_effort
                nexts.append((next_i, next_j))
        return nexts

    def BFS(self, heights):
        next_layer = [(0, 0)]
        while next_layer:
            cur_layer = next_layer
            next_layer = []
            for i, j in cur_layer:
                next_layer.extend(self.findNext(i, j, heights))
        return

    def minimumEffortPath(self, heights: List[List[int]]) -> int:
        self.m = len(heights)
        self.n = len(heights[0])
        self.efforts = [[-1 for _ in range(self.n)] for _ in range(self.m)]
        self.efforts[0][0] = 0
        self.BFS(heights)
        return self.efforts[-1][-1]
```

```python
# 1631 Dijkstra
# time 684ms, beat 88.56%
# space 16.3MB, beat 80.72%
import heapq
class Solution:
    def BFS(self, heights):
        next_layer = []
        heapq.heappush(next_layer, (0, 0, 0))
        while next_layer:
            effort, i, j = heapq.heappop(next_layer)
            if self.visited[i][j]:
                continue
            self.visited[i][j] = True
            if i == self.m - 1 and j == self.n - 1:
                return effort
            for next_i, next_j in [[i-1, j], [i+1, j], [i, j - 1], [i, j + 1]]:
                if next_i < 0 or next_i >= self.m or next_j < 0 or next_j >= self.n or self.visited[next_i][next_j]:
                    continue
                cur_effort = max(effort, abs(heights[next_i][next_j] - heights[i][j]))
                heapq.heappush(next_layer, (cur_effort, next_i, next_j))
        return -1

    def minimumEffortPath(self, heights: List[List[int]]) -> int:
        self.m = len(heights)
        self.n = len(heights[0])
        self.visited = [[False for _ in range(self.n)] for _ in range(self.m)]
        return self.BFS(heights)
```

```python
# 1631 union-find
# time 776ms, beat 74.51%
# space 19MB, beat 21.89%

import heapq
class UnionSet:
    def __init__(self, n):
        self.root = {}
        for i in range(n):
            self.root[i] = i
    def findRoot(self, i):
        while i != self.root[i]:
            i = self.root[i]
        return i

    def add(self, i, j):
        i = self.findRoot(i)
        j = self.findRoot(j)
        if i > j:
            i, j = j, i
        self.root[j] = self.root[i]
    
    def connected(self, i, j):
        i = self.findRoot(i)
        j = self.findRoot(j)
        return j == i

class Solution:
    def minimumEffortPath(self, heights: List[List[int]]) -> int:
        # build a union set for each node
        m = len(heights)
        n = len(heights[0])
        if m * n == 1:
            return 0
        union_set = UnionSet(m*n)
        # add all the edges into a heap
        candidates = []
        for i in range(m):
            for j in range(n):
                for next_i, next_j in [[i+1, j], [i, j+1]]:
                    if next_i >= m or next_j >= n:
                        continue
                    effort = abs(heights[next_i][next_j] - heights[i][j])
                    heapq.heappush(candidates, (effort, next_i * n + next_j, i * n + j))
        # each time pop an edge from the heap and add it to the union set.
        while candidates:
            best_cand = heapq.heappop(candidates)
            union_set.add(best_cand[1], best_cand[2])   
            # if the start and end are in the same set, return the effort
            if union_set.connected(0, m * n - 1):
                return best_cand[0]
        return -1
```

```python
# 1631 binary search
# time 2832ms, beat 21.57%
# space 15.8MB, beat 97.39%

import heapq
class Solution:
    def isReachable(self, mid, heights):
        visited = [[False for _ in range(self.n)] for _ in range(self.m)]
        next_layer = [[0, 0]]
        visited[0][0] = True
        while next_layer:
            cur_layer = next_layer
            next_layer = []
            for i, j in cur_layer:
                if i == self.m - 1 and j == self.n - 1:
                    return True
                for next_i, next_j in [[i - 1, j], [i + 1, j], [i, j - 1], [i, j + 1]]:
                    if next_i < 0 or next_i >= self.m or next_j < 0 or next_j >= self.n or visited[next_i][next_j] or abs(heights[i][j] - heights[next_i][next_j]) > mid:
                        continue
                    visited[next_i][next_j] = True
                    next_layer.append([next_i, next_j])
        return False

    def minimumEffortPath(self, heights: List[List[int]]) -> int:
        # binary search for the possible height
        self.m = len(heights)
        self.n = len(heights[0])
        # print("m: ", self.m, " n: ", self.n)
        right = max([max(l) for l in heights])
        left = 0
        # print(left, right)
        while left < right:
            mid = (left + right) // 2
            if self.isReachable(mid, heights):
                right = mid
            else:
                left = mid + 1
        return right
```

    - [X] If the graph is sparse, which method should you choose? If it's dense, which method? Dense: iterate from the points; sparse: iterate from the edges. The overall idea is to start from whichever side has the lower traversal cost.
    - Approach 1: Dijkstra's algorithm, always picking the smallest edge.
        - Store directly as a list of edges.
            - Complexity analysis: the outer loop over to_check runs O(E) times (the number of edges), and each heap pop costs O(log E). Finding neighbors within the loop happens at most O(N) times (the number of points), since each point's shortest-path value is only finalized once -- after that it's never expanded again to find neighbors. Finding neighbors via a list has complexity O(E), so overall the time complexity is O((E-N) log E + NE log E) = O(NE log E).
            - Space complexity is O(max(E, N)): the O(E) to_check is a heap holding candidate edges to visit, and the O(N) min_path records each point's shortest-path value.
        - What if we use an adjacency list for storage instead?
            - Time complexity: initially building the adjacency list costs O(E); the outer loop complexity is unchanged, but finding neighbors for each point drops overall from O(NE) to O(E), since at most there are N points plus E neighbor entries. Overall it's O(E + (E-N) log E + E log E) = O(E log E)
            - Space complexity: the extra node2children storage costs O(E); everything else is the same as the previous implementation
        - Think carefully about this approach and you'll notice it actually 're-checks' a point's path, just expressed through pushes and pops on the heap. For example: suppose the weight from start to A is 5, from start to B is 1, and from B to A is 1. Then the edge start->A with weight 5 gets pushed first; after processing B, an edge B->A with weight 2 also gets pushed. When popping, weight 2 pops before weight 5 -- but 5 still got counted in the loop. The number of loop iterations equals the number of edges, whereas the steps that are actually 'useful' are only the ones that finalize a point's shortest path, i.e. the number of points. Can we make the traversal count equal to the number of points instead? Yes -- change the loop to, each time, look for the 'minimum point'. Consider the difference between the two approaches.
    - Approach 2: Dijkstra's algorithm, iterating from points, always finding the minimum.
        - Time complexity: building the neighbor lists costs O(E); the outer loop runs N times, and finding the minimum point each time costs O(N); updating neighbors across all points totals O(E). The overall complexity is O(N^2 + E).
        - Space complexity: the extra node2children storage costs O(E), and the O(N) min_path records each point's shortest-path value. Overall it's O(max(E, N)).
The difference between approach 2 and approach 1: the most time-consuming step in approach 1 is popping edges from the heap, which costs O(log E). But for edges that turn out to be discarded -- i.e. edges whose starting point's shortest path is already finalized -- if there are many such edges, that's wasted overhead. Comparing approach 1 and approach 2: when is E log E greater than N^2 + E? When E is large but N is small, i.e. the graph is dense, approach 2 is better. Conversely, if the graph is sparse, approach 1's overhead is smaller, so approach 1 is better.
    - Approach 3: Bellman-Ford algorithm (an improvement on BFS). The idea: since popping from the heap is so costly, don't keep everything in a heap -- just process it directly.
        - Method: Dijkstra's improvement over BFS is that instead of processing every possible candidate each time, it picks out the 'best' candidate
        - The Bellman-Ford algorithm instead adds all possible points on a length-k shortest path (after deduplication) into BFS's next-layer candidates.
        - Time complexity: O(NE) -- you iterate N times, and each time examine at most E edges.
        - Correctness of Bellman-Ford: prove it by induction on the length of the shortest path. It trivially holds for length 0. Assume it holds for length k. Then for a shortest path of length k+1, it must have a length-k subpath d_k(i), and we know d_k(i) + d(i, j) >= d_{k+1}(j). Since i is the second-to-last node on the length-(k+1) shortest path ending at j, i can always 'relax' the original length-(k+1) shortest path ending at j, giving d_{k+1}(j) >= d_k(i) + d(i, j). This proves the claim. [Reference link](https://blog.csdn.net/zhangxiaojiakele/article/details/104951367)

    - This problem has two approaches: front-to-back, and back-to-front.
        - Front-to-back: start from courses that need no prerequisites. Each time a course is finished, remove it from the prerequisite list of any course that depends on it, meaning those courses now have one fewer prerequisite requirement. If there's currently no course without prerequisites, exit the loop. Check whether the number of completed courses equals the total.
        - Back-to-front: start from the last course(s) -- courses that aren't a prerequisite for any other course -- then work backward through their prerequisites, removing each from its prerequisites' list of dependents. If there's currently no course without dependents, end the loop.
        Within the loop, there are two ways to choose the next point: BFS or DFS.
        - BFS: finish all currently takeable courses first, then consider the courses that become takeable afterward.
        - DFS: finish the current course, then look at whether any of its follow-up (dependent) courses are now takeable; if so, pick one to take next; if not, backtrack.

```python
# 210's four implementations (front/back, BFS/DFS)
# front + BFS
# time 44ms, beat 70.84%
# space 16MB, beat 54.79%

# front + DFS
# time 44ms, beat 70.84%
# space 17.8MB, beat 30.30%

# back + BFS
# time 44ms, beat 70.84%
# space 16.2MB, beat 45.46%

# back + DFS
# time 48ms, beat 47.53%
# space 17.9MB, beat 22.91%

from collections import defaultdict
class Solution:
    def ForwardBFS(self, start_nodes):
        next_layer = start_nodes
        res = []
        while next_layer:
            cur_layer = next_layer
            next_layer = []
            for course in cur_layer:
                res.append(course)
                later_courses = self.course2later.get(course, [])
                for later_c in later_courses:
                    self.in_degree[later_c] -= 1
                    if self.in_degree[later_c] == 0:
                        next_layer.append(later_c)
        return res

    def ForwardDFS(self, start):
        self.res.append(start)
        for later_c in self.course2later.get(start, []):
            self.in_degree[later_c] -= 1
            if self.in_degree[later_c] == 0:
                self.ForwardDFS(later_c)

    def BackwordBFS(self, ends):
        res = []
        next_layer = ends
        while next_layer:
            cur_layer = next_layer
            next_layer = []
            for course in cur_layer:
                res.append(course)
                for pre in self.course2pre.get(course, []):
                    self.out_degree[pre] -= 1
                    if self.out_degree[pre] == 0:
                        next_layer.append(pre)
        return res[::-1]

    def BackwordDFS(self, node):
        self.res.append(node)
        for pre in self.course2pre.get(node, []):
            self.out_degree[pre] -= 1
            if self.out_degree[pre] == 0:
                self.BackwordDFS(pre)


    def findOrder(self, numCourses: int, prerequisites: List[List[int]]) -> List[int]:
        direction = "F" # "B"
        search = 'B' # or "D"
        if direction == "F":
            self.in_degree = [0 for i in range(numCourses)]
            self.course2later = defaultdict(list)
            for pair in prerequisites:
                self.in_degree[pair[0]] += 1
                self.course2later[pair[1]].append(pair[0])
            starts = []
            for i in range(numCourses):
                if self.in_degree[i] == 0:
                    starts.append(i)
            if not starts:
                return []
            if search == "B":
                res = self.ForwardBFS(starts)
                return res if len(res) == numCourses else []
            else:
                self.res = []
                for start in starts:
                    self.ForwardDFS(start)
                return self.res if len(self.res) == numCourses else []
        else:
            self.out_degree = [0 for _ in range(numCourses)]
            self.course2pre = defaultdict(list)
            for pair in prerequisites:
                self.out_degree[pair[1]] += 1
                self.course2pre[pair[0]].append(pair[1])
            ends = []
            for i in range(numCourses):
                if self.out_degree[i] == 0:
                    ends.append(i)
            if not ends:
                return []
            if search == "B":
                res = self.BackwordBFS(ends)
                return res if len(res) == numCourses else []
            else:
                self.res = []
                for end in ends:
                    self.BackwordDFS(end)
            return self.res[::-1] if len(self.res) == numCourses else []
```
